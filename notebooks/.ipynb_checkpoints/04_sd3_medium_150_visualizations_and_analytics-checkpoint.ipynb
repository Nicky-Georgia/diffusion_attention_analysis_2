```python

def read_json(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def sample_evenly(items, n):
    items = list(items)
    if len(items) <= n:
        return items
    idxs = np.linspace(0, len(items) - 1, n).round().astype(int).tolist()
    seen = []
    for i in idxs:
        if i not in seen:
            seen.append(i)
    return [items[i] for i in seen]


def open_image(path: Path, size=(256, 256)) -> Image.Image:
    img = Image.open(path).convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def open_image_from_pil(img: Image.Image, size=(256, 256)) -> Image.Image:
    img = img.convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def make_grid(images, titles=None, ncols=4, cell_size=(256, 296), title_h=40, out_path=None):
    images = list(images)
    if not images:
        return None
    titles = titles or [""] * len(images)
    nrows = math.ceil(len(images) / ncols)
    grid = Image.new("RGB", (ncols * cell_size[0], nrows * cell_size[1]), "white")
    draw = ImageDraw.Draw(grid)
    for idx, (img, title) in enumerate(zip(images, titles)):
        row, col = divmod(idx, ncols)
        x = col * cell_size[0]
        y = row * cell_size[1]
        if img.size != (cell_size[0], cell_size[0]):
            img = open_image_from_pil(img, size=(cell_size[0], cell_size[0]))
        grid.paste(img, (x, y + title_h))
        draw.text((x + 8, y + 8), str(title)[:56], fill=(0, 0, 0))
    if out_path is not None:
        grid.save(out_path)
    return grid


def display_saved(path: Path):
    if path.exists():
        display(IPyImage(filename=str(path)))
    else:
        print("Missing:", path)


def safe_stage_report(stage: str, path: Path):
    data = read_json(path, {}) or {}
    return {
        "stage": stage,
        "exists": path.exists(),
        "status": data.get("status"),
        "samples": data.get("samples"),
        "rows": data.get("rows"),
        "labels": data.get("labels"),
        "masks": data.get("masks"),
        "path": str(path),
    }


def du_sh(path: Path) -> str:
    if not path.exists() and not path.is_symlink():
        return "missing"
    try:
        out = subprocess.check_output(["du", "-sh", str(path)], text=True, stderr=subprocess.DEVNULL)
        return out.split()[0]
    except Exception:
        return "?"
```

## 3. CLIPSeg mask overlays used by Stage 4

```python

def overlay_mask(image_path: Path, mask_path: Path, size=(256, 256)) -> Image.Image:
    base = Image.open(image_path).convert("RGB").resize(size)
    mask = Image.open(mask_path).convert("L").resize(size)
    base_arr = np.asarray(base).astype(np.float32)
    mask_arr = (np.asarray(mask).astype(np.float32) / 255.0)[..., None]
    overlay = base_arr.copy()
    overlay[..., 0] = np.clip(overlay[..., 0] * (1 - 0.45 * mask_arr[..., 0]) + 255 * 0.45 * mask_arr[..., 0], 0, 255)
    overlay[..., 1] = overlay[..., 1] * (1 - 0.25 * mask_arr[..., 0])
    overlay[..., 2] = overlay[..., 2] * (1 - 0.25 * mask_arr[..., 0])
    return Image.fromarray(overlay.astype(np.uint8))

manifest_path = MASK_DIR / "mask_manifest.csv"
if manifest_path.exists():
    mask_df = pd.read_csv(manifest_path)
    print("mask rows:", len(mask_df), "samples:", mask_df["sample_id"].nunique())
    display(
        mask_df.groupby("label")
        .agg(samples=("sample_id", "nunique"), mean_area=("mask_area_fraction", "mean"))
        .sort_values("samples", ascending=False)
        .head(25)
    )
    show_rows = mask_df.sort_values("mask_area_fraction", ascending=False).groupby("label").head(1).head(12)
    imgs, titles = [], []
    for _, row in show_rows.iterrows():
        image_path = CAPTURE_DIR / "samples" / row["sample_id"] / "image.png"
        mask_path = Path(row["mask_path"])
        if image_path.exists() and mask_path.exists():
            imgs.append(overlay_mask(image_path, mask_path))
            titles.append(f"{row['label']}\narea={row['mask_area_fraction']:.2f}")
    out = VIS_DIR / "02_sd3_clipseg_mask_overlays.png"
    make_grid(imgs, titles, ncols=4, out_path=out)
    display_saved(out)
else:
    print("Mask manifest is missing:", manifest_path)
```

## 6. Stage 5 same-seed intervention difference analytics

```python

def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }


def collect_intervention_diffs(stage_dir: Path, *, image_name: str, kind: str) -> pd.DataFrame:
    rows = read_json(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id", "")
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = CAPTURE_DIR / "samples" / base_id / "image.png"
        edited_path = stage_dir / "samples" / sample_id / image_name
        if not baseline_path.exists() or not edited_path.exists():
            continue
        item = {
            "kind": kind,
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "baseline_path": str(baseline_path),
            "edited_path": str(edited_path),
        }
        item.update(_image_diff_metrics(baseline_path, edited_path))
        out.append(item)
    return pd.DataFrame(out)

if STAGE5_DIFFS_CSV.exists():
    all_diffs = pd.read_csv(STAGE5_DIFFS_CSV)
else:
    tables = []
    if BLOCK_SCALE_DIR.exists():
        tables.append(collect_intervention_diffs(BLOCK_SCALE_DIR, image_name="ablation.png", kind="block_scaling_strong"))
    if SAE_LABEL_PROBE_ROOT.exists():
        for label_dir in sorted(SAE_LABEL_PROBE_ROOT.iterdir()):
            if label_dir.is_dir():
                tables.append(collect_intervention_diffs(label_dir, image_name="steered.png", kind=f"sae_label/{label_dir.name}"))
    all_diffs = pd.concat([x for x in tables if len(x)], ignore_index=True) if tables else pd.DataFrame()
    if len(all_diffs):
        STAGE5_DIFFS_CSV.parent.mkdir(parents=True, exist_ok=True)
        all_diffs.to_csv(STAGE5_DIFFS_CSV, index=False)

if len(all_diffs):
    display(
        all_diffs.groupby("kind")
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_max=("l1_mean", "max"),
            rmse_mean=("rmse", "mean"),
            changed_px_010=("changed_px_010", "mean"),
        )
        .sort_values("l1_mean", ascending=False)
    )
    display(all_diffs.sort_values("l1_mean", ascending=False).head(30)[["kind", "sample_id", "beta", "scale", "l1_mean", "rmse", "changed_px_010"]])
else:
    print("No Stage 5 diff rows found. Run the patched Stage 5 cells first.")
```

## 7. Stage 5 representative examples

```python

if len(all_diffs):
    top = all_diffs.sort_values("l1_mean", ascending=False).head(12)
    imgs, titles = [], []
    for _, row in top.iterrows():
        baseline_path = Path(row["baseline_path"])
        edited_path = Path(row.get("edited_path") or row.get("image_path"))
        if not baseline_path.exists() or not edited_path.exists():
            continue
        imgs.append(open_image(baseline_path, size=(224, 224)))
        imgs.append(open_image(edited_path, size=(224, 224)))
        param = ""
        if not pd.isna(row.get("beta", np.nan)):
            param = f"beta={row['beta']}"
        elif not pd.isna(row.get("scale", np.nan)):
            param = f"scale={row['scale']}"
        titles.append(f"baseline\n{str(row['base_id'])[:34]}")
        titles.append(f"{row['kind']} {param}\nL1={row['l1_mean']:.3f}")
    out = VIS_DIR / "06_stage5_most_changed_examples.png"
    make_grid(imgs, titles, ncols=4, cell_size=(224, 264), title_h=40, out_path=out)
    display_saved(out)
else:
    print("No Stage 5 examples to visualize.")
```

```python
def read_json(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def open_image(path: Path, size=(256, 256)) -> Image.Image:
    img = Image.open(path).convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def open_image_from_pil(img: Image.Image, size=(256, 256)) -> Image.Image:
    img = img.convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def diff_image(a_path: Path, b_path: Path, size=(256, 256)) -> Image.Image:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    d = np.abs(y - x).mean(axis=2)
    out = np.zeros((size[1], size[0], 3), dtype=np.uint8)
    out[..., 0] = np.clip(d * 255 * 2.0, 0, 255).astype(np.uint8)
    out[..., 1] = np.clip(d * 255 * 0.7, 0, 180).astype(np.uint8)
    out[..., 2] = np.clip(d * 255 * 0.4, 0, 120).astype(np.uint8)
    return Image.fromarray(out)


def make_grid(images, titles=None, ncols=4, cell_size=(224, 264), title_h=40, out_path=None):
    images = list(images)
    if not images:
        return None
    titles = titles or [""] * len(images)
    nrows = math.ceil(len(images) / ncols)
    grid = Image.new("RGB", (ncols * cell_size[0], nrows * cell_size[1]), "white")
    draw = ImageDraw.Draw(grid)
    for idx, (img, title) in enumerate(zip(images, titles)):
        row, col = divmod(idx, ncols)
        x = col * cell_size[0]
        y = row * cell_size[1]
        if img.size != (cell_size[0], cell_size[0]):
            img = open_image_from_pil(img, size=(cell_size[0], cell_size[0]))
        grid.paste(img, (x, y + title_h))
        draw.text((x + 6, y + 6), str(title)[:58], fill=(0, 0, 0))
    if out_path is not None:
        grid.save(out_path)
    return grid


def display_saved(path: Path):
    if path.exists():
        display(IPyImage(filename=str(path)))
    else:
        print("Missing:", path)


def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _find_baseline_image(base_id: str) -> Path | None:
    for p in [
        CAPTURE_DIR / "samples" / base_id / "image.png",
        SPATIAL_CAPTURE_DIR / "samples" / base_id / "image.png",
        ORIGINAL_CAPTURE_DIR / "samples" / base_id / "image.png",
    ]:
        if p.exists():
            return p
    return None


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }
```

```python
def collect_stage5_v2_diffs(stage_dir: Path, *, image_name: str, kind: str, label: str | None = None, module: str | None = None, window: str | None = None) -> pd.DataFrame:
    rows = read_json(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = _find_baseline_image(base_id)
        edited_path = stage_dir / "samples" / sample_id / image_name
        if baseline_path is None or not edited_path.exists():
            continue
        item = {
            "kind": kind,
            "label": label or row.get("target_word", ""),
            "module": module or (row.get("requested_modules") or [None])[0],
            "window": window,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "target_word": row.get("target_word", label or ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "baseline_path": str(baseline_path),
            "edited_path": str(edited_path),
        }
        item.update(_image_diff_metrics(baseline_path, edited_path))
        out.append(item)
    return pd.DataFrame(out)

if STAGE5_V2_DIFFS_CSV.exists():
    all_diffs_v2 = pd.read_csv(STAGE5_V2_DIFFS_CSV)
else:
    tables = []
    if BLOCK_SOFT_ROOT.exists():
        for d in sorted([p for p in BLOCK_SOFT_ROOT.iterdir() if p.is_dir()]):
            name = d.name
            if "__" in name:
                module_part, window = name.rsplit("__", 1)
                module = module_part.replace("transformer_blocks_", "transformer_blocks.")
            else:
                module, window = None, None
            tables.append(collect_stage5_v2_diffs(d, image_name="ablation.png", kind="block_soft_v2", module=module, window=window))
    if SAE_SPECIFIC_ROOT.exists():
        for label_dir in sorted([p for p in SAE_SPECIFIC_ROOT.iterdir() if p.is_dir()]):
            for window_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
                tables.append(collect_stage5_v2_diffs(window_dir, image_name="steered.png", kind="sae_specificity_v2", label=label_dir.name, window=window_dir.name))
    if SAE_TEMPORAL_ROOT.exists():
        for label_dir in sorted([p for p in SAE_TEMPORAL_ROOT.iterdir() if p.is_dir()]):
            for window_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
                tables.append(collect_stage5_v2_diffs(window_dir, image_name="steered.png", kind="sae_temporal_v2", label=label_dir.name, window=window_dir.name))
    all_diffs_v2 = pd.concat([x for x in tables if len(x)], ignore_index=True) if tables else pd.DataFrame()
    if len(all_diffs_v2):
        all_diffs_v2["artifact_flag"] = (all_diffs_v2["l1_mean"] > 0.35) | (all_diffs_v2["changed_px_020"] > 0.70)
        all_diffs_v2["moderate_change"] = (all_diffs_v2["l1_mean"].between(0.03, 0.25)) & (~all_diffs_v2["artifact_flag"])
        all_diffs_v2.to_csv(STAGE5_V2_DIFFS_CSV, index=False)

if len(all_diffs_v2):
    for col in ["beta", "scale", "l1_mean", "changed_px_010", "changed_px_020"]:
        if col in all_diffs_v2.columns:
            all_diffs_v2[col] = pd.to_numeric(all_diffs_v2[col], errors="coerce")
    if "artifact_flag" not in all_diffs_v2.columns:
        all_diffs_v2["artifact_flag"] = (all_diffs_v2["l1_mean"] > 0.35) | (all_diffs_v2["changed_px_020"] > 0.70)
    if "moderate_change" not in all_diffs_v2.columns:
        all_diffs_v2["moderate_change"] = (all_diffs_v2["l1_mean"].between(0.03, 0.25)) & (~all_diffs_v2["artifact_flag"])
    print("Stage 5 v2 rows:", len(all_diffs_v2))
    display(all_diffs_v2.head())
else:
    print("No Stage 5 v2 diffs found. Run the Stage 5 improved patch first.")
```

```python
from pathlib import Path
import math
import textwrap
import numpy as np
import pandas as pd
from PIL import Image, ImageDraw, ImageFont

def _load_unicode_font(size: int = 12):
    """Prefer a TrueType font with Unicode support; fall back safely if unavailable."""
    candidates = [
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
        "/usr/share/fonts/truetype/dejavu/DejaVuSansCondensed.ttf",
        "/usr/share/fonts/truetype/liberation2/LiberationSans-Regular.ttf",
        "/usr/share/fonts/truetype/freefont/FreeSans.ttf",
        "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc",
        "/System/Library/Fonts/Supplemental/Arial Unicode.ttf",
        "/Library/Fonts/Arial Unicode.ttf",
    ]
    for path in candidates:
        p = Path(path)
        if p.exists():
            try:
                return ImageFont.truetype(str(p), size=size), True, str(p)
            except Exception:
                pass
    return ImageFont.load_default(), False, "PIL default latin-1 font"

_GRID_FONT, _GRID_FONT_UNICODE, _GRID_FONT_NAME = _load_unicode_font(size=12)
print("Grid font:", _GRID_FONT_NAME)

def _safe_title_text(text, *, width: int = 32, max_chars: int = 140) -> str:
    """Wrap title text and make it safe for the currently loaded PIL font."""
    text = "" if text is None else str(text)

    # If TrueType Unicode font is not available, replace common scientific symbols.
    if not _GRID_FONT_UNICODE:
        replacements = {
            "β": "beta",
            "α": "alpha",
            "γ": "gamma",
            "Δ": "Delta",
            "δ": "delta",
            "λ": "lambda",
            "μ": "mu",
            "×": "x",
            "→": "->",
            "←": "<-",
            "≤": "<=",
            "≥": ">=",
            "±": "+/-",
            "–": "-",
            "—": "-",
            "…": "...",
        }
        for src, dst in replacements.items():
            text = text.replace(src, dst)
        text = text.encode("latin-1", errors="replace").decode("latin-1")

    wrapped = []
    for line in text.splitlines():
        line = line.strip()
        if not line:
            wrapped.append("")
        else:
            wrapped.extend(textwrap.wrap(line, width=width) or [""])
    return "\n".join(wrapped)[:max_chars]

def _fit_pil_image(img: Image.Image, size=(224, 224), bg=(255, 255, 255)) -> Image.Image:
    """Resize/pad PIL image to fixed canvas."""
    img = img.convert("RGB")
    img.thumbnail(size, Image.LANCZOS)
    canvas = Image.new("RGB", size, bg)
    x = (size[0] - img.width) // 2
    y = (size[1] - img.height) // 2
    canvas.paste(img, (x, y))
    return canvas

def make_grid(images, titles=None, ncols=3, cell_size=(224, 264), title_h=40, out_path=None):
    """Unicode-safe image grid renderer.

    Supports PIL images and paths. Uses notebook's existing open_image_from_pil/open_image
    helpers when available.
    """
    images = list(images)
    titles = list(titles or [""] * len(images))

    if len(images) == 0:
        print("make_grid: no images to render.")
        return None

    ncols = max(1, int(ncols))
    nrows = int(math.ceil(len(images) / ncols))
    cell_w, cell_h = int(cell_size[0]), int(cell_size[1])
    img_size = (cell_w, max(1, cell_h - title_h))
    grid = Image.new("RGB", (ncols * cell_w, nrows * cell_h), (255, 255, 255))
    draw = ImageDraw.Draw(grid)

    for i, img in enumerate(images):
        x = (i % ncols) * cell_w
        y = (i // ncols) * cell_h

        if isinstance(img, Image.Image):
            if "open_image_from_pil" in globals():
                try:
                    img = open_image_from_pil(img, size=img_size)
                except Exception:
                    img = _fit_pil_image(img, size=img_size)
            else:
                img = _fit_pil_image(img, size=img_size)
        else:
            if "open_image" not in globals():
                raise RuntimeError("Expected helper function open_image(path, size=...) to be defined.")
            img = open_image(Path(img), size=img_size)

        grid.paste(img, (x, y + title_h))
        title = titles[i] if i < len(titles) else ""
        draw.multiline_text(
            (x + 6, y + 5),
            _safe_title_text(title),
            fill=(0, 0, 0),
            font=_GRID_FONT,
            spacing=1,
        )

    if out_path is not None:
        out_path = Path(out_path)
        out_path.parent.mkdir(parents=True, exist_ok=True)
        grid.save(out_path)
    return grid

def _fmt_num(x, default=""):
    try:
        if pd.isna(x):
            return default
        return f"{float(x):g}"
    except Exception:
        return default

def make_triplet_gallery(rows: pd.DataFrame, out_path: Path, *, title_prefix: str, max_rows: int = 8):
    """Render baseline / edited / diff triplets without Unicode font crashes."""
    imgs, titles = [], []

    if rows is None or len(rows) == 0:
        print(f"No rows for gallery: {out_path}")
        return None

    for _, row in rows.head(max_rows).iterrows():
        base = Path(row["baseline_path"])
        edit = Path(row["edited_path"])
        if not base.exists() or not edit.exists():
            continue

        beta = row.get("beta", np.nan)
        scale = row.get("scale", np.nan)
        if not pd.isna(beta):
            param = f"beta={_fmt_num(beta)}"
        elif not pd.isna(scale):
            param = f"scale={_fmt_num(scale)}"
        else:
            param = ""

        l1 = row.get("l1_mean", np.nan)
        px020 = row.get("changed_px_020", np.nan)
        base_id = str(row.get("base_id", ""))

        imgs.extend([
            open_image(base, size=(224, 224)),
            open_image(edit, size=(224, 224)),
            diff_image(base, edit, size=(224, 224)),
        ])
        titles.extend([
            f"baseline\n{base_id[:28]}",
            f"{str(title_prefix).replace('β', 'beta')} {param}\nL1={_fmt_num(l1)}",
            f"diff\npx>.20={_fmt_num(px020)}",
        ])

    if not imgs:
        print(f"No existing image pairs for gallery: {out_path}")
        return None

    grid = make_grid(imgs, titles, ncols=3, cell_size=(224, 264), title_h=40, out_path=out_path)
    if out_path is not None and Path(out_path).exists() and "display_saved" in globals():
        display_saved(out_path)
    else:
        display(grid)
    return grid
```

```python
def make_triplet_gallery(rows: pd.DataFrame, out_path: Path, *, title_prefix: str, max_rows: int = 8):
    imgs, titles = [], []
    for _, row in rows.head(max_rows).iterrows():
        base = Path(row["baseline_path"])
        edit = Path(row["edited_path"])
        if not base.exists() or not edit.exists():
            continue
        param = ""
        if not pd.isna(row.get("beta", np.nan)):
            param = f"β={row['beta']}"
        elif not pd.isna(row.get("scale", np.nan)):
            param = f"scale={row['scale']}"
        imgs.extend([
            open_image(base, size=(224, 224)),
            open_image(edit, size=(224, 224)),
            diff_image(base, edit, size=(224, 224)),
        ])
        titles.extend([
            f"baseline\n{row['base_id'][:28]}",
            f"{title_prefix} {param}\nL1={row['l1_mean']:.3f}",
            f"diff\npx>.20={row['changed_px_020']:.2f}",
        ])
    make_grid(imgs, titles, ncols=3, cell_size=(224, 264), title_h=40, out_path=out_path)
    display_saved(out_path)

if len(all_diffs_v2):
    # 1) Useful moderate SAE cases.
    sae_mod = all_diffs_v2[(all_diffs_v2["kind"] == "sae_specificity_v2") & (all_diffs_v2["moderate_change"])]
    if len(sae_mod):
        # Prefer strongest moderate changes while avoiding artifacts.
        rows = sae_mod.sort_values(["l1_mean", "changed_px_020"], ascending=[False, True])
        out = VIS_DIR / "stage5_v2_sae_moderate_non_artifact_triplets.png"
        make_triplet_gallery(rows, out, title_prefix="SAE")
    else:
        print("No moderate SAE candidates by current heuristic. Showing lowest-artifact SAE examples instead.")
        rows = all_diffs_v2[all_diffs_v2["kind"] == "sae_specificity_v2"].sort_values(["artifact_flag", "changed_px_020", "l1_mean"], ascending=[True, True, False])
        out = VIS_DIR / "stage5_v2_sae_low_artifact_triplets.png"
        make_triplet_gallery(rows, out, title_prefix="SAE")

    # 2) Block cases around median effect, not just max L1.
    block = all_diffs_v2[all_diffs_v2["kind"] == "block_soft_v2"].copy()
    if len(block):
        target = block["l1_mean"].median()
        block["median_distance"] = (block["l1_mean"] - target).abs()
        rows = block.sort_values(["artifact_flag", "median_distance"]).head(8)
        out = VIS_DIR / "stage5_v2_block_median_triplets.png"
        make_triplet_gallery(rows, out, title_prefix="block")

    # 3) Sanity beta=0.
    beta0 = all_diffs_v2[(all_diffs_v2["kind"] == "sae_specificity_v2") & (all_diffs_v2["beta"] == 0)]
    if len(beta0):
        rows = beta0.sort_values("l1_mean", ascending=False)
        out = VIS_DIR / "stage5_v2_sae_beta0_sanity_triplets.png"
        make_triplet_gallery(rows, out, title_prefix="SAE β0", max_rows=4)
else:
    print("No Stage 5 v2 rows for galleries.")
```

```python
RUN_CLIPSEG_STAGE5_V2_EVAL = True
MAX_CLIPSEG_ROWS = 80
CLIPSEG_THRESHOLD = 0.35
CLIPSEG_EVAL_CSV = STAGE5_V2_ROOT / "stage5_v2_clipseg_semantic_eval.csv"


def _clipseg_centroid(prob: np.ndarray):
    prob = np.asarray(prob, dtype=np.float64)
    total = prob.sum()
    if total <= 1e-8:
        return np.nan, np.nan
    yy, xx = np.mgrid[0:prob.shape[0], 0:prob.shape[1]]
    return float((xx * prob).sum() / total / max(prob.shape[1] - 1, 1)), float((yy * prob).sum() / total / max(prob.shape[0] - 1, 1))


def _clipseg_stats(prob: np.ndarray, threshold: float = CLIPSEG_THRESHOLD):
    cx, cy = _clipseg_centroid(prob)
    return {
        "score_mean": float(prob.mean()),
        "score_p90": float(np.quantile(prob, 0.90)),
        "score_max": float(prob.max()),
        "area_thr": float((prob >= threshold).mean()),
        "centroid_x": cx,
        "centroid_y": cy,
    }

if RUN_CLIPSEG_STAGE5_V2_EVAL and len(all_diffs_v2):
    import torch
    from transformers import CLIPSegForImageSegmentation, CLIPSegProcessor

    sae_rows = all_diffs_v2[(all_diffs_v2["kind"] == "sae_specificity_v2") & (all_diffs_v2["beta"] != 0)].copy()
    # Prefer non-artifact/moderate rows first; cap to keep runtime manageable.
    sae_rows["priority"] = np.where(sae_rows["moderate_change"], 0, np.where(~sae_rows["artifact_flag"], 1, 2))
    sae_rows = sae_rows.sort_values(["priority", "label", "beta", "l1_mean"]).head(MAX_CLIPSEG_ROWS)

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model_id = "CIDAS/clipseg-rd64-refined"
    print("Loading CLIPSeg:", model_id, "device:", device, "rows:", len(sae_rows))
    processor = CLIPSegProcessor.from_pretrained(model_id)
    model = CLIPSegForImageSegmentation.from_pretrained(model_id).to(device).eval()

    eval_rows = []
    for idx, row in sae_rows.iterrows():
        label = str(row.get("label") or row.get("target_word") or "").strip().lower()
        if not label:
            continue
        images = [Image.open(row["baseline_path"]).convert("RGB"), Image.open(row["edited_path"]).convert("RGB")]
        with torch.no_grad():
            inputs = processor(text=[label, label], images=images, return_tensors="pt", padding=True)
            inputs = {k: (v.to(device) if hasattr(v, "to") else v) for k, v in inputs.items()}
            logits = model(**inputs).logits
            if logits.ndim == 2:
                logits = logits.unsqueeze(0)
            probs = torch.sigmoid(logits).detach().cpu().numpy()
        base_stats = _clipseg_stats(probs[0])
        edit_stats = _clipseg_stats(probs[1])
        cx_shift = np.nan
        cy_shift = np.nan
        if not np.isnan(base_stats["centroid_x"]) and not np.isnan(edit_stats["centroid_x"]):
            cx_shift = edit_stats["centroid_x"] - base_stats["centroid_x"]
            cy_shift = edit_stats["centroid_y"] - base_stats["centroid_y"]
        eval_rows.append({
            **row.to_dict(),
            "clipseg_label": label,
            "base_score_mean": base_stats["score_mean"],
            "edit_score_mean": edit_stats["score_mean"],
            "delta_score_mean": edit_stats["score_mean"] - base_stats["score_mean"],
            "base_score_p90": base_stats["score_p90"],
            "edit_score_p90": edit_stats["score_p90"],
            "delta_score_p90": edit_stats["score_p90"] - base_stats["score_p90"],
            "base_area_thr": base_stats["area_thr"],
            "edit_area_thr": edit_stats["area_thr"],
            "delta_area_thr": edit_stats["area_thr"] - base_stats["area_thr"],
            "centroid_shift_x": cx_shift,
            "centroid_shift_y": cy_shift,
            "centroid_shift_l2": float(np.sqrt(cx_shift ** 2 + cy_shift ** 2)) if not np.isnan(cx_shift) else np.nan,
        })
        if len(eval_rows) % 10 == 0:
            print("processed", len(eval_rows), "/", len(sae_rows))

    clipseg_eval_df = pd.DataFrame(eval_rows)
    clipseg_eval_df.to_csv(CLIPSEG_EVAL_CSV, index=False)
    print("Saved:", CLIPSEG_EVAL_CSV)

elif CLIPSEG_EVAL_CSV.exists():
    clipseg_eval_df = pd.read_csv(CLIPSEG_EVAL_CSV)
    print("Loaded existing:", CLIPSEG_EVAL_CSV)
else:
    clipseg_eval_df = pd.DataFrame()
    print("CLIPSeg eval skipped or no rows.")

if len(clipseg_eval_df):
    summary = (
        clipseg_eval_df.groupby(["label", "beta"], dropna=False)
        .agg(
            n=("delta_score_mean", "count"),
            delta_score_mean=("delta_score_mean", "mean"),
            delta_score_p90=("delta_score_p90", "mean"),
            delta_area_thr=("delta_area_thr", "mean"),
            centroid_shift_l2=("centroid_shift_l2", "mean"),
            l1_mean=("l1_mean", "mean"),
            artifact_rate=("artifact_flag", "mean"),
        )
        .reset_index()
        .sort_values(["delta_score_mean", "delta_score_p90"], ascending=False)
    )
    display(summary)

    # Gallery of best target-score improvements that are not flagged as artifacts.
    best = clipseg_eval_df[~clipseg_eval_df["artifact_flag"]].sort_values("delta_score_mean", ascending=False).head(8)
    if len(best):
        imgs, titles = [], []
        for _, row in best.iterrows():
            base = Path(row["baseline_path"])
            edit = Path(row["edited_path"])
            imgs.extend([open_image(base, size=(224, 224)), open_image(edit, size=(224, 224)), diff_image(base, edit, size=(224, 224))])
            titles.extend([
                f"baseline {row['label']}\nscore={row['base_score_mean']:.3f}",
                f"edited β={row['beta']}\nΔscore={row['delta_score_mean']:+.3f}",
                f"diff\nL1={row['l1_mean']:.3f}",
            ])
        out = VIS_DIR / "stage5_v2_clipseg_best_semantic_triplets.png"
        make_grid(imgs, titles, ncols=3, cell_size=(224, 264), title_h=40, out_path=out)
        display_saved(out)
```
