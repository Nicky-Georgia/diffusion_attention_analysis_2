## 2. Source hotfix v3: image-token extraction + active-token additive SAE steering


```python
PATCH_STAGE5_V3_SOURCE = True


def _patch_between(path: Path, start_marker: str, end_marker: str, new_text: str, patch_marker: str, *, allow_replace_existing: bool = False):
    text = path.read_text(encoding="utf-8")
    if patch_marker in text and not allow_replace_existing:
        print(f"Already patched: {path.name} ({patch_marker})")
        return False
    start = text.index(start_marker)
    end = text.index(end_marker, start)
    backup = path.with_suffix(path.suffix + f".before_{patch_marker.lower()}")
    if not backup.exists():
        backup.write_text(text, encoding="utf-8")
    path.write_text(text[:start] + new_text.rstrip() + text[end:], encoding="utf-8")
    print("Patched:", path)
    print("Backup:", backup)
    return True

if PATCH_STAGE5_V3_SOURCE:
    act_path = PROJECT_ROOT / "src" / "diffusion_attention_analysis_v2" / "activations.py"
    act_text = act_path.read_text(encoding="utf-8")
    if "SD3_TUPLE_SPATIAL_HOTFIX" not in act_text:
        new_extract = """def extract_first_tensor(obj: Any):
    # SD3_TUPLE_SPATIAL_HOTFIX:
    # SD3 JointTransformerBlock tuples can contain (encoder_hidden_states, hidden_states).
    # The larger token stream is the image-token stream needed for spatial analysis/interventions.
    if isinstance(obj, torch.Tensor):
        return obj
    if isinstance(obj, (tuple, list)):
        direct_tensors = [item for item in obj if isinstance(item, torch.Tensor)]
        if len(direct_tensors) >= 2:
            a, b = direct_tensors[0], direct_tensors[1]
            if a.ndim == 3 and b.ndim == 3 and b.shape[1] > a.shape[1]:
                return b
        for item in obj:
            found = extract_first_tensor(item)
            if found is not None:
                return found
    if isinstance(obj, dict):
        for value in obj.values():
            found = extract_first_tensor(value)
            if found is not None:
                return found
    return None
"""
        _patch_between(act_path, "def extract_first_tensor", "\n\ndef replace_first_tensor", new_extract, "SD3_TUPLE_SPATIAL_HOTFIX")
    else:
        print("extract_first_tensor already has SD3 spatial hotfix")

    act_text = act_path.read_text(encoding="utf-8")
    if "SD3_RECURSIVE_REPLACE_HOTFIX" not in act_text:
        new_replace = """def replace_first_tensor(obj: Any, new_tensor: torch.Tensor, old_shape) -> Any:
    # SD3_RECURSIVE_REPLACE_HOTFIX:
    # Recursively replace the first tensor matching old_shape.
    def _replace(x: Any):
        if isinstance(x, torch.Tensor):
            if tuple(x.shape) == tuple(old_shape):
                return new_tensor, True
            return x, False
        if isinstance(x, tuple):
            out = []
            changed_any = False
            for item in x:
                if changed_any:
                    out.append(item)
                    continue
                new_item, changed = _replace(item)
                out.append(new_item)
                changed_any = changed_any or changed
            return tuple(out), changed_any
        if isinstance(x, list):
            out = []
            changed_any = False
            for item in x:
                if changed_any:
                    out.append(item)
                    continue
                new_item, changed = _replace(item)
                out.append(new_item)
                changed_any = changed_any or changed
            return out, changed_any
        if isinstance(x, dict):
            out = {}
            changed_any = False
            for key, value in x.items():
                if changed_any:
                    out[key] = value
                    continue
                new_value, changed = _replace(value)
                out[key] = new_value
                changed_any = changed_any or changed
            return out, changed_any
        return x, False

    out, _ = _replace(obj)
    return out
"""
        _patch_between(act_path, "def replace_first_tensor", "\n\ndef flatten_feature_map", new_replace, "SD3_RECURSIVE_REPLACE_HOTFIX")
    else:
        print("replace_first_tensor already has recursive hotfix")

    int_path = PROJECT_ROOT / "src" / "diffusion_attention_analysis_v2" / "interventions.py"
    new_class = """class ResidualConceptSteerer:
    # SD3_ACTIVE_TOKEN_STEERING_V3:
    # Add SAE decoder directions to the original residual update.
    # Optional active-token mode localizes the edit to spatial tokens where the selected SAE features already activate.
    def __init__(self, sae, concept_ids: Sequence[int], beta: float, *, step_getter: Callable[[], int], active_steps: Sequence[int] | None = None, device: str = "cuda") -> None:
        self.sae = sae.to(device)
        self.concept_ids = [int(x) for x in concept_ids]
        self.beta = float(beta)
        self.step_getter = step_getter
        self.active_steps = {int(x) for x in active_steps or []}
        self.device = device
        import os as _os
        self.branch_mode = _os.environ.get("SD3_SAE_BRANCH_MODE", "cond").strip().lower()
        self.direction_mode = _os.environ.get("SD3_SAE_DIRECTION_MODE", "sum_decoder_direction").strip().lower()
        self.spatial_mode = _os.environ.get("SD3_SAE_SPATIAL_MODE", "all").strip().lower()
        self.top_frac = float(_os.environ.get("SD3_SAE_TOP_FRAC", "0.15"))
        self.active_quantile = float(_os.environ.get("SD3_SAE_ACTIVE_QUANTILE", "0.85"))
        self.outside_scale = float(_os.environ.get("SD3_SAE_OUTSIDE_SCALE", "0.0"))
        self.normalize_direction = _os.environ.get("SD3_SAE_NORMALIZE_DIRECTION", "none").strip().lower()
        self.beta_scale_mode = _os.environ.get("SD3_SAE_BETA_SCALE_MODE", "constant").strip().lower()

    def _valid_ids(self, max_dim: int) -> list[int]:
        return [i for i in self.concept_ids if 0 <= int(i) < int(max_dim)]

    def _direction(self, flat: torch.Tensor) -> torch.Tensor | None:
        with torch.no_grad():
            W = self.sae.decoder.weight.detach().to(device=flat.device, dtype=flat.dtype)
            valid = self._valid_ids(W.shape[1])
            if not valid:
                return None
            dirs = W[:, valid]
            if self.direction_mode in {"mean", "mean_decoder_direction"}:
                direction = dirs.mean(dim=1)
            else:
                direction = dirs.sum(dim=1)
            if self.normalize_direction in {"rms", "unit_rms"}:
                direction = direction / direction.pow(2).mean().sqrt().clamp_min(1e-6)
            elif self.normalize_direction in {"l2", "unit_l2"}:
                direction = direction / direction.norm().clamp_min(1e-6)
            return direction.reshape(1, -1)

    def _branch_mask(self, feature_shape: Sequence[int], flat: torch.Tensor) -> torch.Tensor:
        mask = torch.ones((flat.shape[0], 1), device=flat.device, dtype=flat.dtype)
        if len(feature_shape) != 3:
            return mask
        b, n, c = [int(x) for x in feature_shape]
        if b < 2 or b % 2 != 0:
            return mask
        branch_mode = self.branch_mode if self.branch_mode in {"cond", "conditional", "uncond", "unconditional", "all"} else "cond"
        if branch_mode == "all":
            return mask
        m = torch.zeros((b, n, 1), device=flat.device, dtype=flat.dtype)
        half = b // 2
        if branch_mode in {"cond", "conditional"}:
            m[half:] = 1.0
        else:
            m[:half] = 1.0
        return m.reshape(-1, 1)

    def _active_token_mask(self, flat: torch.Tensor, feature_shape: Sequence[int]) -> torch.Tensor:
        if self.spatial_mode in {"all", "global", "none"} or len(feature_shape) != 3:
            return torch.ones((flat.shape[0], 1), device=flat.device, dtype=flat.dtype)
        b, n, c = [int(x) for x in feature_shape]
        with torch.no_grad():
            z = self.sae.encode(flat)
            valid = self._valid_ids(z.shape[-1])
            if not valid:
                return torch.ones((flat.shape[0], 1), device=flat.device, dtype=flat.dtype)
            signal = z[:, valid].float().amax(dim=-1).reshape(b, n)
            mask = torch.full((b, n), float(self.outside_scale), device=flat.device, dtype=flat.dtype)
            if self.spatial_mode in {"top_frac", "active_top", "active_tokens"}:
                k = max(1, min(n, int(round(float(self.top_frac) * n))))
                idx = torch.topk(signal, k=k, dim=1).indices
                mask.scatter_(1, idx, 1.0)
            else:
                q = float(self.active_quantile)
                q = min(max(q, 0.0), 0.999)
                thresh = torch.quantile(signal, q, dim=1, keepdim=True)
                active = signal >= thresh
                # Fallback if activations are nearly flat.
                need_fallback = active.sum(dim=1) < 1
                if need_fallback.any():
                    k = max(1, min(n, int(round(float(self.top_frac) * n))))
                    idx = torch.topk(signal[need_fallback], k=k, dim=1).indices
                    rows = torch.where(need_fallback)[0]
                    active[rows] = False
                    active[rows.unsqueeze(1), idx] = True
                mask[active] = 1.0
            return mask.reshape(-1, 1).to(dtype=flat.dtype)

    def __call__(self, module, inputs, output):
        step = int(self.step_getter())
        if self.active_steps and step not in self.active_steps:
            return output
        x_in = extract_first_tensor(inputs)
        x_out = extract_first_tensor(output)
        if x_out is None:
            return output
        residual_mode = x_in is not None and tuple(x_in.shape) == tuple(x_out.shape)
        update = x_out - x_in if residual_mode else x_out
        try:
            flat, shape = flatten_feature_map(update.detach().float())
        except ValueError:
            return output
        flat = flat.to(self.device)
        direction = self._direction(flat)
        if direction is None:
            return output
        with torch.no_grad():
            if self.beta_scale_mode in {"activation_rms", "token_rms"}:
                local_scale = flat.pow(2).mean(dim=-1, keepdim=True).sqrt().clamp_min(1e-6)
                delta = self.beta * local_scale * direction.expand_as(flat)
            else:
                delta = self.beta * direction.expand_as(flat)
            delta = delta * self._active_token_mask(flat, shape)
            delta = delta * self._branch_mask(shape, flat)
            steered_flat = flat + delta
        steered = unflatten_feature_map(steered_flat, shape, like=x_out)
        new_tensor = x_in + steered if residual_mode else steered
        return replace_first_tensor(output, new_tensor, x_out.shape)
"""
    _patch_between(
        int_path,
        "class ResidualConceptSteerer:",
        "\n\nclass BlockOutputScaler:",
        new_class,
        "SD3_ACTIVE_TOKEN_STEERING_V3",
        allow_replace_existing=True,
    )

print("Source hotfix v3 completed")
```

## 3. Feature selection v3


```python
def _safe_file_part(text: str) -> str:
    text = str(text).strip().lower()
    text = re.sub(r"[^a-z0-9_-]+", "_", text)
    text = re.sub(r"_+", "_", text).strip("_")
    return text or "label"


def _yaml_list(xs):
    return "[" + ",".join(str(x) for x in xs) + "]"


def _read_jsonl(path: Path) -> list[dict]:
    rows = []
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


def _target_word_from_record(row: dict) -> str:
    if row.get("target_word"):
        return str(row["target_word"]).strip().lower()
    for key in ("target", "label"):
        if row.get(key):
            return str(row[key]).strip().lower()
    entities = row.get("entities") or []
    if entities:
        ent = entities[0]
        if isinstance(ent, dict):
            return str(ent.get("token") or ent.get("label") or ent.get("name") or "").strip().lower()
        return str(ent).strip().lower()
    return ""


MAX_STAGE5_LABELS_V3 = 6
PROBE_PROMPTS_PER_LABEL_V3 = 2
FEATURES_PER_LABEL_V3 = 3

PREFERRED_LABELS_V3 = ["person", "dog", "car", "book", "bicycle", "cat", "cup", "chair", "vase", "bird", "tree"]

concept_df = pd.read_csv(CONCEPT_DIR / "concept_dictionary.csv")
concept_df["label"] = concept_df["label"].astype(str).str.lower()
concept_df["feature_id"] = pd.to_numeric(concept_df["feature_id"], errors="coerce").astype("Int64")
concept_df["score"] = pd.to_numeric(concept_df.get("score", 0.0), errors="coerce").fillna(0.0)
concept_df = concept_df.dropna(subset=["feature_id"]).copy()
concept_df["feature_id"] = concept_df["feature_id"].astype(int)

fl = (
    concept_df.groupby(["feature_id", "label"], as_index=False)
    .agg(score=("score", "max"), count=("count", "max"))
)
feature_totals = fl.groupby("feature_id").agg(
    n_labels=("label", "nunique"),
    sum_score=("score", "sum"),
    max_score_any=("score", "max"),
).reset_index()
fl = fl.merge(feature_totals, on="feature_id", how="left")
fl["label_share"] = fl["score"] / fl["sum_score"].clip(lower=1e-8)

other_rows = []
for fid, grp in fl.groupby("feature_id"):
    srt = grp.sort_values("score", ascending=False).reset_index(drop=True)
    for idx, row in srt.iterrows():
        if len(srt) == 1:
            other_max = 0.0
        elif idx == 0:
            other_max = float(srt.loc[1, "score"])
        else:
            other_max = float(srt.loc[0, "score"])
        other_rows.append({"feature_id": int(fid), "label": row["label"], "other_max_score": other_max, "rank_for_feature": idx + 1})
fl = fl.merge(pd.DataFrame(other_rows), on=["feature_id", "label"], how="left")
fl["specificity_margin"] = fl["score"] - fl["other_max_score"].fillna(0.0)
fl["selectivity_score"] = fl["score"] * fl["label_share"] / np.sqrt(fl["n_labels"].clip(lower=1))

fl["tier"] = "reject"
fl.loc[(fl["n_labels"] <= 6) & ((fl["specificity_margin"] >= 1.0) | (fl["label_share"] >= 0.16)), "tier"] = "strict"
fl.loc[(fl["tier"] == "reject") & (fl["n_labels"] <= 12) & ((fl["specificity_margin"] >= 0.20) | (fl["label_share"] >= 0.09)), "tier"] = "balanced"
fl.loc[(fl["tier"] == "reject") & (fl["n_labels"] <= 18) & (fl["rank_for_feature"] <= 5) & (fl["score"] >= fl["score"].quantile(0.55)), "tier"] = "strong_non_ultrabroad"

fl.to_csv(STAGE5_V3_CANDIDATES_CSV, index=False)

prompt_records_all = _read_jsonl(PROMPTS_150)
for r in prompt_records_all:
    r["target_word"] = _target_word_from_record(r)
target_counts = pd.Series([r["target_word"] for r in prompt_records_all if r.get("target_word")]).value_counts().to_dict()
concept_labels = set(concept_df["label"].unique())


def select_features(label: str, mode: str, k: int = FEATURES_PER_LABEL_V3) -> pd.DataFrame:
    rows = fl[fl["label"] == label].copy()
    if not len(rows):
        return rows
    if mode == "strict":
        rows = rows[rows["tier"] == "strict"]
    elif mode == "balanced":
        rows = rows[rows["tier"].isin(["strict", "balanced"])]
    elif mode == "strong":
        rows = rows[rows["tier"].isin(["strict", "balanced", "strong_non_ultrabroad"])]
    else:
        raise ValueError(mode)
    if not len(rows):
        return rows
    tier_order = {"strict": 0, "balanced": 1, "strong_non_ultrabroad": 2, "reject": 3}
    rows["tier_order"] = rows["tier"].map(tier_order).fillna(9)
    rows = rows.sort_values(
        ["tier_order", "selectivity_score", "specificity_margin", "score"],
        ascending=[True, False, False, False],
    )
    return rows.head(k)

candidate_labels = []
for label in PREFERRED_LABELS_V3:
    if target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL_V3 and label in concept_labels and len(select_features(label, "strong")):
        candidate_labels.append(label)
    if len(candidate_labels) >= MAX_STAGE5_LABELS_V3:
        break

if len(candidate_labels) < MAX_STAGE5_LABELS_V3:
    ranked = (
        fl[fl["tier"].isin(["strict", "balanced", "strong_non_ultrabroad"])]
        .groupby("label", as_index=False)
        .agg(best_selectivity=("selectivity_score", "max"), best_score=("score", "max"))
        .sort_values(["best_selectivity", "best_score"], ascending=False)
    )
    for label in ranked["label"]:
        if label not in candidate_labels and target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL_V3:
            candidate_labels.append(label)
        if len(candidate_labels) >= MAX_STAGE5_LABELS_V3:
            break

TARGET_LABELS_V3 = candidate_labels[:MAX_STAGE5_LABELS_V3]
assert TARGET_LABELS_V3, "No usable labels for Stage 5 v3"
print("Target labels v3:", TARGET_LABELS_V3)

# Write per-label prompt files.
PROBE_PROMPT_ROOT_V3.mkdir(parents=True, exist_ok=True)
probe_label_prompt_paths = {}
probe_records = []
for label in TARGET_LABELS_V3:
    records = [dict(r) for r in prompt_records_all if r.get("target_word") == label][:PROBE_PROMPTS_PER_LABEL_V3]
    for r in records:
        r["target_word"] = label
    path = PROBE_PROMPT_ROOT_V3 / f"sd3_v3_probe_{_safe_file_part(label)}.jsonl"
    with path.open("w", encoding="utf-8") as f:
        for r in records:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")
    probe_label_prompt_paths[label] = path
    probe_records.extend(records)

BLOCK_PROMPTS_V3 = PROBE_PROMPT_ROOT_V3 / "sd3_v3_block_probe.jsonl"
with BLOCK_PROMPTS_V3.open("w", encoding="utf-8") as f:
    for r in probe_records[: min(8, len(probe_records))]:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

selection_rows = []
label_feature_sets = {"balanced": {}, "strong": {}}
for label in TARGET_LABELS_V3:
    for mode in ["balanced", "strong"]:
        selected = select_features(label, mode)
        ids = selected["feature_id"].astype(int).tolist()
        label_feature_sets[mode][label] = ids
        for _, row in selected.iterrows():
            selection_rows.append({
                "label": label,
                "mode": mode,
                "feature_id": int(row["feature_id"]),
                "score": float(row["score"]),
                "n_labels": int(row["n_labels"]),
                "label_share": float(row["label_share"]),
                "specificity_margin": float(row["specificity_margin"]),
                "selectivity_score": float(row["selectivity_score"]),
                "tier": str(row["tier"]),
                "rank_for_feature": int(row["rank_for_feature"]),
            })
selection_df = pd.DataFrame(selection_rows)
selection_df.to_csv(STAGE5_V3_SELECTION_CSV, index=False)

display(selection_df)
print("Feature candidates saved:", STAGE5_V3_CANDIDATES_CSV)
print("Feature selection saved:", STAGE5_V3_SELECTION_CSV)
print("Block prompt file:", BLOCK_PROMPTS_V3)
print("Prompt files:")
for k, v in probe_label_prompt_paths.items():
    print(" ", k, "->", v)
```

## 5c. SAE steering v3 — active-token localized mode

This is the main new attempt. Instead of adding the SAE direction to all `64×64` tokens, the hook computes current SAE activations and applies the direction only to the most active tokens for the selected concept IDs. This should improve locality and allow stronger beta without global artifacts.


```python
RUN_SAE_ACTIVE_TOKENS_V3 = True

SAE_ACTIVE_V3_WINDOWS = {
    "mid": [11, 12, 13, 14, 15],
    "early_bridge": [6, 7, 8, 9, 10, 11, 12, 13],
}
SAE_ACTIVE_V3_RUN_WINDOWS = ["mid", "early_bridge"]
SAE_ACTIVE_V3_BETAS = [-40.0, 0.0, 40.0, 60.0]
SAE_ACTIVE_V3_FEATURE_MODE = "strong"  # localized mode can tolerate a slightly stronger feature bundle.
SAE_ACTIVE_V3_MAX_PROMPTS_PER_LABEL = PROBE_PROMPTS_PER_LABEL_V3

# Active-token steering env.
os.environ["SD3_SAE_BRANCH_MODE"] = "cond"
os.environ["SD3_SAE_DIRECTION_MODE"] = "sum_decoder_direction"
os.environ["SD3_SAE_SPATIAL_MODE"] = "active_quantile"
os.environ["SD3_SAE_ACTIVE_QUANTILE"] = "0.85"
os.environ["SD3_SAE_TOP_FRAC"] = "0.15"
os.environ["SD3_SAE_OUTSIDE_SCALE"] = "0.0"
os.environ["SD3_SAE_NORMALIZE_DIRECTION"] = "none"
os.environ["SD3_SAE_BETA_SCALE_MODE"] = "constant"

sae_active_reports = []
for label in TARGET_LABELS_V3:
    concept_ids = label_feature_sets[SAE_ACTIVE_V3_FEATURE_MODE].get(label, [])
    prompt_file = probe_label_prompt_paths.get(label)
    if not concept_ids or prompt_file is None:
        print("[skip]", label, concept_ids, prompt_file)
        continue
    for window_name in SAE_ACTIVE_V3_RUN_WINDOWS:
        target_steps = SAE_ACTIVE_V3_WINDOWS[window_name]
        out_dir = SAE_ACTIVE_V3_ROOT / _safe_file_part(label) / SAE_ACTIVE_V3_FEATURE_MODE / window_name
        overrides = [
            f"data.prompts_path={prompt_file}",
            f"data.sae_checkpoint={SAE_DIR / 'sae.pt'}",
            f"data.output_dir={out_dir}",
            f"data.max_prompts={SAE_ACTIVE_V3_MAX_PROMPTS_PER_LABEL}",
            f"runtime.seed_start={SEED_START}",
            f"runtime.num_seeds={NUM_SEEDS}",
            f"intervention.module={SAE_LAYER}",
            f"intervention.concept_ids={_yaml_list(concept_ids)}",
            f"intervention.betas={_yaml_list(SAE_ACTIVE_V3_BETAS)}",
            f"intervention.target_steps={_yaml_list(target_steps)}",
        ]
        print(f"\n=== SAE active-token v3 | label={label} | ids={concept_ids} | window={window_name} ===")
        run_config(str(SD3_SAE_CFG_PATH_V3), overrides=overrides, dry_run=True)
        if RUN_SAE_ACTIVE_TOKENS_V3:
            run_config(str(SD3_SAE_CFG_PATH_V3), overrides=overrides, dry_run=False)
            report = show_report(str(out_dir / "report.json")) or {}
            sae_active_reports.append({"label": label, "mode": SAE_ACTIVE_V3_FEATURE_MODE, "window": window_name, "concept_ids": concept_ids, "out_dir": str(out_dir), **report})

if sae_active_reports:
    display(pd.DataFrame(sae_active_reports))
else:
    print("RUN_SAE_ACTIVE_TOKENS_V3=False: dry-runs only")
```

## 6. Stage 5 v3 diffs and viz


```python
from PIL import Image, ImageDraw, ImageFont, ImageChops


def _read_json_safe(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _find_baseline_image(base_id: str) -> Path | None:
    candidates = [
        SPATIAL_CAPTURE_DIR / "samples" / base_id / "image.png",
        ORIGINAL_CAPTURE_DIR / "samples" / base_id / "image.png",
    ]
    for p in candidates:
        if p.exists():
            return p
    return None


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }


def collect_diffs(stage_dir: Path, *, image_name: str, kind: str, label: str | None = None, module: str | None = None, window: str | None = None, mode: str | None = None) -> pd.DataFrame:
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = _find_baseline_image(base_id)
        edited_path = stage_dir / "samples" / sample_id / image_name
        if baseline_path is None or not edited_path.exists():
            continue
        meta = {
            "kind": kind,
            "label": label or row.get("target_word", ""),
            "module": module or (row.get("requested_modules") or [None])[0],
            "window": window,
            "mode": mode,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "baseline_path": str(baseline_path),
            "edited_path": str(edited_path),
        }
        meta.update(_image_diff_metrics(baseline_path, edited_path))
        out.append(meta)
    return pd.DataFrame(out)


diff_tables = []
if BLOCK_V3_ROOT.exists():
    for d in sorted([p for p in BLOCK_V3_ROOT.iterdir() if p.is_dir()]):
        name = d.name
        if "__" in name:
            module_part, window = name.rsplit("__", 1)
            module = module_part.replace("transformer_blocks_", "transformer_blocks.")
        else:
            module, window = None, None
        diff_tables.append(collect_diffs(d, image_name="ablation.png", kind="block_v3", module=module, window=window))

for root, kind in [(SAE_GLOBAL_V3_ROOT, "sae_global_v3"), (SAE_ACTIVE_V3_ROOT, "sae_active_tokens_v3")]:
    if root.exists():
        for label_dir in sorted([p for p in root.iterdir() if p.is_dir()]):
            for mode_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
                for window_dir in sorted([p for p in mode_dir.iterdir() if p.is_dir()]):
                    diff_tables.append(collect_diffs(window_dir, image_name="steered.png", kind=kind, label=label_dir.name, mode=mode_dir.name, window=window_dir.name))

all_diffs_v3 = pd.concat([d for d in diff_tables if len(d)], ignore_index=True) if diff_tables else pd.DataFrame()
if len(all_diffs_v3):
    all_diffs_v3["artifact_flag"] = (all_diffs_v3["l1_mean"] > 0.35) | (all_diffs_v3["changed_px_020"] > 0.70)
    all_diffs_v3["moderate_change"] = (all_diffs_v3["l1_mean"].between(0.03, 0.30)) & (~all_diffs_v3["artifact_flag"])
    all_diffs_v3.to_csv(STAGE5_V3_DIFFS_CSV, index=False)
    print("Saved:", STAGE5_V3_DIFFS_CSV)

    print("Overall summary:")
    display(
        all_diffs_v3.groupby(["kind", "window"], dropna=False)
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_median=("l1_mean", "median"),
            changed_px_010=("changed_px_010", "mean"),
            changed_px_020=("changed_px_020", "mean"),
            artifact_rate=("artifact_flag", "mean"),
            moderate_rate=("moderate_change", "mean"),
        )
        .reset_index()
        .sort_values("l1_mean", ascending=False)
    )

    print("SAE by label / mode / window / beta:")
    display(
        all_diffs_v3[all_diffs_v3["kind"].str.startswith("sae")]
        .groupby(["kind", "label", "mode", "window", "beta"], dropna=False)
        .agg(n=("l1_mean", "count"), l1_mean=("l1_mean", "mean"), changed_px_020=("changed_px_020", "mean"), artifact_rate=("artifact_flag", "mean"), moderate_rate=("moderate_change", "mean"))
        .reset_index()
        .sort_values(["artifact_rate", "l1_mean"], ascending=[True, False])
        .head(60)
    )

    print("Most useful non-artifact candidates:")
    display(
        all_diffs_v3[(~all_diffs_v3["artifact_flag"]) & (all_diffs_v3["l1_mean"] > 0.02)]
        .sort_values(["kind", "l1_mean"], ascending=[True, False])
        .head(40)[["kind", "label", "mode", "window", "module", "beta", "scale", "l1_mean", "changed_px_010", "changed_px_020", "sample_id"]]
    )
else:
    print("No Stage 5 v3 images found yet. Run the Stage 5 cells first.")
```

```python
VIS_V3_DIR = PROJECT_ROOT / "outputs" / RUN_TAG / "visualizations" / "stage5_v3_retry"
VIS_V3_DIR.mkdir(parents=True, exist_ok=True)


def _load_font(size=14):
    for path in [
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
        "/usr/local/share/fonts/DejaVuSans.ttf",
    ]:
        if Path(path).exists():
            return ImageFont.truetype(path, size=size)
    return ImageFont.load_default()

FONT = _load_font(13)


def open_image(path: Path, size=(224, 224)):
    return Image.open(path).convert("RGB").resize(size)


def diff_image(a_path: Path, b_path: Path, size=(224, 224)):
    a = open_image(a_path, size)
    b = open_image(b_path, size)
    d = ImageChops.difference(a, b)
    # amplify for visibility
    arr = np.asarray(d).astype(np.float32)
    arr = np.clip(arr * 3.0, 0, 255).astype(np.uint8)
    return Image.fromarray(arr)


def make_grid(images, titles, ncols=3, cell_size=(224, 264), title_h=40, out_path=None):
    if not images:
        print("No images for grid", out_path)
        return None
    nrows = math.ceil(len(images) / ncols)
    W, H = cell_size
    grid = Image.new("RGB", (ncols * W, nrows * H), "white")
    draw = ImageDraw.Draw(grid)
    for idx, img in enumerate(images):
        r, c = divmod(idx, ncols)
        x, y = c * W, r * H
        grid.paste(img, (x, y + title_h))
        draw.multiline_text((x + 6, y + 5), str(titles[idx])[:90], fill=(0, 0, 0), font=FONT, spacing=2)
    if out_path:
        grid.save(out_path)
        display(grid)
        print("Saved:", out_path)
    return grid


def make_triplet_gallery(rows: pd.DataFrame, out_path: Path, *, title_prefix: str, max_rows: int = 8):
    imgs, titles = [], []
    for _, row in rows.head(max_rows).iterrows():
        base = Path(row["baseline_path"])
        edit = Path(row["edited_path"])
        if not base.exists() or not edit.exists():
            continue
        if not pd.isna(row.get("beta", np.nan)):
            param = f"beta={float(row['beta']):g}"
        elif not pd.isna(row.get("scale", np.nan)):
            param = f"scale={float(row['scale']):g}"
        else:
            param = ""
        imgs.extend([open_image(base), open_image(edit), diff_image(base, edit)])
        titles.extend([
            f"baseline\n{row['base_id'][:26]}",
            f"{title_prefix} {param}\nL1={row['l1_mean']:.3f}",
            f"diff\npx>.20={row['changed_px_020']:.2f}",
        ])
    return make_grid(imgs, titles, ncols=3, out_path=out_path)

if len(all_diffs_v3):
    sae_good = all_diffs_v3[(all_diffs_v3["kind"].str.startswith("sae")) & (~all_diffs_v3["artifact_flag"]) & (all_diffs_v3["l1_mean"] > 0.02)]
    if len(sae_good):
        rows = sae_good.sort_values(["moderate_change", "l1_mean"], ascending=[False, False])
        make_triplet_gallery(rows, VIS_V3_DIR / "stage5_v3_sae_best_non_artifact_triplets.png", title_prefix="SAE v3")
    else:
        print("No non-artifact SAE rows with l1_mean > 0.02")

    active_good = all_diffs_v3[(all_diffs_v3["kind"] == "sae_active_tokens_v3") & (~all_diffs_v3["artifact_flag"])]
    if len(active_good):
        rows = active_good.sort_values("l1_mean", ascending=False)
        make_triplet_gallery(rows, VIS_V3_DIR / "stage5_v3_active_tokens_triplets.png", title_prefix="active")

    block_good = all_diffs_v3[(all_diffs_v3["kind"] == "block_v3") & (~all_diffs_v3["artifact_flag"])]
    if len(block_good):
        rows = block_good.sort_values("l1_mean", ascending=False)
        make_triplet_gallery(rows, VIS_V3_DIR / "stage5_v3_block_triplets.png", title_prefix="block")
else:
    print("No Stage 5 v3 diffs available for galleries.")
```

## 7. Optional semantic check with CLIPSeg

Run only after v3 images are generated. This is slower and may download/load CLIPSeg, so it is disabled by default. It measures whether the target-label CLIPSeg score changes in the edited image.


```python
RUN_CLIPSEG_STAGE5_V3_EVAL = True
CLIPSEG_MAX_ROWS = 80

if RUN_CLIPSEG_STAGE5_V3_EVAL and len(all_diffs_v3):
    import torch
    from transformers import CLIPSegProcessor, CLIPSegForImageSegmentation

    device = "cuda" if torch.cuda.is_available() else "cpu"
    processor = CLIPSegProcessor.from_pretrained("CIDAS/clipseg-rd64-refined")
    model = CLIPSegForImageSegmentation.from_pretrained("CIDAS/clipseg-rd64-refined").to(device).eval()

    def clipseg_score(image_path: Path, label: str) -> float:
        image = Image.open(image_path).convert("RGB")
        inputs = processor(text=[label], images=[image], padding=True, return_tensors="pt").to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
            prob = torch.sigmoid(logits).float().mean().item()
        return float(prob)

    eval_rows = []
    rows = all_diffs_v3[all_diffs_v3["kind"].str.startswith("sae")].copy()
    rows = rows[(~rows["artifact_flag"]) & (rows["label"].astype(str).str.len() > 0)].sort_values("l1_mean", ascending=False).head(CLIPSEG_MAX_ROWS)
    for _, row in rows.iterrows():
        label = str(row["label"])
        b = clipseg_score(Path(row["baseline_path"]), label)
        e = clipseg_score(Path(row["edited_path"]), label)
        eval_rows.append({**row.to_dict(), "clipseg_before": b, "clipseg_after": e, "clipseg_delta": e - b})
    clipseg_v3 = pd.DataFrame(eval_rows)
    out_csv = STAGE5_V3_ROOT / "stage5_v3_clipseg_eval.csv"
    clipseg_v3.to_csv(out_csv, index=False)
    print("Saved:", out_csv)
    display(
        clipseg_v3.groupby(["kind", "label", "mode", "window", "beta"], dropna=False)
        .agg(n=("clipseg_delta", "count"), delta_mean=("clipseg_delta", "mean"), before=("clipseg_before", "mean"), after=("clipseg_after", "mean"), l1_mean=("l1_mean", "mean"))
        .reset_index()
        .sort_values("delta_mean", ascending=False)
        .head(40)
    )
else:
    print("CLIPSeg v3 eval skipped. Set RUN_CLIPSEG_STAGE5_V3_EVAL=True after image generation if needed.")
```
