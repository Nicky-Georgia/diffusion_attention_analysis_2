```python
def read_json(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def sample_evenly(items, n):
    items = list(items)
    if len(items) <= n:
        return items
    idxs = np.linspace(0, len(items) - 1, n).round().astype(int).tolist()
    seen = []
    for i in idxs:
        if i not in seen:
            seen.append(i)
    return [items[i] for i in seen]


def open_image(path: Path, size=(256, 256)) -> Image.Image:
    img = Image.open(path).convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def open_image_from_pil(img: Image.Image, size=(256, 256)) -> Image.Image:
    img = img.convert("RGB")
    img.thumbnail(size)
    canvas = Image.new("RGB", size, "white")
    canvas.paste(img, ((size[0] - img.width) // 2, (size[1] - img.height) // 2))
    return canvas


def make_grid(images, titles=None, ncols=4, cell_size=(256, 296), title_h=40, out_path=None):
    images = list(images)
    if not images:
        return None
    titles = titles or [""] * len(images)
    nrows = math.ceil(len(images) / ncols)
    grid = Image.new("RGB", (ncols * cell_size[0], nrows * cell_size[1]), "white")
    draw = ImageDraw.Draw(grid)
    for idx, (img, title) in enumerate(zip(images, titles)):
        row, col = divmod(idx, ncols)
        x = col * cell_size[0]
        y = row * cell_size[1]
        if img.size != (cell_size[0], cell_size[0]):
            img = open_image_from_pil(img, size=(cell_size[0], cell_size[0]))
        grid.paste(img, (x, y + title_h))
        draw.text((x + 8, y + 8), str(title)[:48], fill=(0, 0, 0))
    if out_path is not None:
        grid.save(out_path)
    return grid


def display_saved(path: Path):
    if path.exists():
        display(IPyImage(filename=str(path)))
    else:
        print("Missing:", path)


def safe_stage_report(stage, path):
    data = read_json(path, {}) or {}
    return {
        "stage": stage,
        "exists": path.exists(),
        "status": data.get("status"),
        "samples": data.get("samples"),
        "rows": data.get("rows"),
        "labels": data.get("labels"),
        "masks": data.get("masks"),
        "path": str(path),
    }


def du_sh(path: Path) -> str:
    if not path.exists() and not path.is_symlink():
        return "missing"
    try:
        out = subprocess.check_output(["du", "-sh", str(path)], text=True, stderr=subprocess.DEVNULL)
        return out.split()[0]
    except Exception:
        return "?"
```

## 4. Representative CLIPSeg mask overlays

```python
def overlay_mask(image_path: Path, mask_path: Path, size=(256, 256), alpha=0.45) -> Image.Image:
    img = Image.open(image_path).convert("RGB")
    mask = Image.open(mask_path).convert("L").resize(img.size, Image.Resampling.NEAREST)
    red = Image.new("RGB", img.size, (255, 0, 0))
    mask_arr = np.array(mask)
    alpha_mask = Image.fromarray((mask_arr > 127).astype(np.uint8) * int(255 * alpha), mode="L")
    out = img.copy()
    out.paste(red, (0, 0), alpha_mask)
    return open_image_from_pil(out, size=size)

mask_manifest = MASK_DIR / "mask_manifest.csv"
if mask_manifest.exists():
    mask_df = pd.read_csv(mask_manifest)
    print("mask rows:", len(mask_df), "samples:", mask_df["sample_id"].nunique())
    label_stats = mask_df.groupby("label").agg(samples=("sample_id", "nunique"), mean_area=("mask_area_fraction", "mean")).sort_values("samples", ascending=False)
    display(label_stats.head(25))

    diverse = mask_df.sort_values("sample_id").groupby("label", as_index=False).head(1)
    chosen_masks = sample_evenly(diverse.to_dict("records"), 12)
    imgs, titles = [], []
    for row in chosen_masks:
        sample_id = row["sample_id"]
        image_path = CAPTURE_DIR / "samples" / sample_id / "image.png"
        mask_path = Path(row["mask_path"])
        if image_path.exists() and mask_path.exists():
            imgs.append(overlay_mask(image_path, mask_path))
            titles.append(f"{sample_id}\n{row['label']}")
    out = VIS_DIR / "02_mask_overlays.png"
    make_grid(imgs, titles, ncols=4, out_path=out)
    display_saved(out)
else:
    print("No mask manifest:", mask_manifest)
```

## 5. One attention heatmap example across denoising steps

```python
def infer_hw(num_queries: int, meta: dict):
    h, w = meta.get("height"), meta.get("width")
    if h and w and int(h) * int(w) == int(num_queries):
        return int(h), int(w)
    side = int(int(num_queries) ** 0.5)
    if side * side == int(num_queries):
        return side, side
    return 1, int(num_queries)


def heatmap_overlay(image_path: Path, heat: np.ndarray, size=(256, 256), alpha=0.45) -> Image.Image:
    import matplotlib.cm as cm
    img = Image.open(image_path).convert("RGB")
    h = heat.astype(np.float32)
    h = h - np.nanmin(h)
    if np.nanmax(h) > 0:
        h = h / np.nanmax(h)
    rgba = cm.viridis(h)
    rgb = (rgba[:, :, :3] * 255).astype(np.uint8)
    hm = Image.fromarray(rgb).resize(img.size, Image.Resampling.BILINEAR)
    out = Image.blend(img, hm, alpha=alpha)
    return open_image_from_pil(out, size=size)

loc_path = LOC_DIR / "attention_localization_rows.csv"
if loc_path.exists() and (CAPTURE_DIR / "captures").exists():
    loc_df = pd.read_csv(loc_path)
    if len(loc_df):
        middle = loc_df[loc_df["step"] == 24]
        ex = (middle if len(middle) else loc_df).sort_values("iou", ascending=False).iloc[0]
        sample_id = ex["sample_id"]
        label = str(ex["label"]).lower()
        image_path = CAPTURE_DIR / "samples" / sample_id / "image.png"
        metadata = read_json(CAPTURE_DIR / "captures" / sample_id / "metadata.json", {}) or {}
        token_indices = metadata.get("token_indices_by_word", {}).get(label, [])
        imgs, titles = [], []
        for step in [0, 24, 49]:
            step_dir = CAPTURE_DIR / "captures" / sample_id / "attention" / f"step_{step:03d}"
            if not step_dir.exists() or not token_indices:
                continue
            files = sorted(step_dir.glob("*.pt"))
            if not files:
                continue
            preferred = [p for p in files if "mid_block" in p.name] or [p for p in files if "up_blocks_1_attentions_0" in p.name] or files
            path = preferred[0]
            try:
                import torch
                payload = torch.load(path, map_location="cpu")
                att = payload["attention"].float()
                if att.ndim == 4:
                    att = att.mean(dim=1)
                branch = att.shape[0] - 1
                arr = att[branch]
                tok = [int(t) for t in token_indices if 0 <= int(t) < arr.shape[-1]]
                if not tok:
                    continue
                score = arr[:, tok].mean(dim=-1).numpy()
                hh, ww = infer_hw(score.shape[0], payload.get("meta", {}))
                heat = score.reshape(hh, ww)
                imgs.append(heatmap_overlay(image_path, heat))
                titles.append(f"step {step} | {label}")
            except Exception as exc:
                print("Could not load attention", path, exc)
        out = VIS_DIR / "03_attention_heatmap_steps.png"
        make_grid(imgs, titles, ncols=max(1, len(imgs)), out_path=out)
        display_saved(out)
        print("example:", sample_id, "label:", label, "row_iou:", ex.get("iou"))
    else:
        print("Localization rows are empty.")
else:
    print("Localization output is missing:", loc_path)
```

### more examples

```python
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image
from IPython.display import display


N_EXAMPLES_PER_BUCKET = 6
HEATMAP_STEPS = [0, 24, 49]
HEATMAP_BUCKETS = ["high_iou", "medium_iou", "low_iou"]


def _pick_diverse_rows(df: pd.DataFrame, n: int, *, mode: str) -> pd.DataFrame:
    df = df.copy()
    if not len(df):
        return df

    if mode == "high_iou":
        df = df.sort_values("iou", ascending=False)
    elif mode == "low_iou":
        df = df.sort_values("iou", ascending=True)
    elif mode == "medium_iou":
        med = df["iou"].median()
        df["distance_to_median"] = (df["iou"] - med).abs()
        df = df.sort_values("distance_to_median", ascending=True)
    else:
        raise ValueError(mode)

    chosen = []
    used_labels = set()
    used_samples = set()

    for _, row in df.iterrows():
        label = str(row["label"])
        sample_id = str(row["sample_id"])
        if label in used_labels or sample_id in used_samples:
            continue
        chosen.append(row)
        used_labels.add(label)
        used_samples.add(sample_id)
        if len(chosen) >= n:
            break

    if len(chosen) < n:
        for _, row in df.iterrows():
            sample_id = str(row["sample_id"])
            if sample_id in used_samples:
                continue
            chosen.append(row)
            used_samples.add(sample_id)
            if len(chosen) >= n:
                break

    return pd.DataFrame(chosen)


def _load_token_heatmap(sample_id: str, label: str, step: int):
    image_path = CAPTURE_DIR / "samples" / sample_id / "image.png"
    metadata = read_json(CAPTURE_DIR / "captures" / sample_id / "metadata.json", {}) or {}
    token_indices = metadata.get("token_indices_by_word", {}).get(str(label).lower(), [])

    if not image_path.exists() or not token_indices:
        return None

    step_dir = CAPTURE_DIR / "captures" / sample_id / "attention" / f"step_{step:03d}"
    if not step_dir.exists():
        return None

    files = sorted(step_dir.glob("*.pt"))
    if not files:
        return None

    preferred = (
        [p for p in files if "mid_block" in p.name]
        or [p for p in files if "up_blocks_1_attentions_0" in p.name]
        or [p for p in files if "down_blocks_2_attentions_1" in p.name]
        or files
    )
    path = preferred[0]

    import torch
    payload = torch.load(path, map_location="cpu")
    att = payload["attention"].float()

    if att.ndim == 4:
        att = att.mean(dim=1)

    branch = att.shape[0] - 1
    arr = att[branch]

    tok = [int(t) for t in token_indices if 0 <= int(t) < arr.shape[-1]]
    if not tok:
        return None

    score = arr[:, tok].mean(dim=-1).numpy()
    hh, ww = infer_hw(score.shape[0], payload.get("meta", {}))
    heat = score.reshape(hh, ww)
    return heatmap_overlay(image_path, heat, size=(224, 224), alpha=0.45)


loc_path = LOC_DIR / "attention_localization_rows.csv"
assert loc_path.exists(), f"Missing localization rows: {loc_path}"

loc_df = pd.read_csv(loc_path)
mid_df = loc_df[loc_df["step"] == 24].copy()
if not len(mid_df):
    mid_df = loc_df.copy()

for bucket in HEATMAP_BUCKETS:
    chosen = _pick_diverse_rows(mid_df, N_EXAMPLES_PER_BUCKET, mode=bucket)
    imgs, titles = [], []

    for _, row in chosen.iterrows():
        sample_id = str(row["sample_id"])
        label = str(row["label"])
        image_path = CAPTURE_DIR / "samples" / sample_id / "image.png"

        if image_path.exists():
            imgs.append(open_image(image_path, size=(224, 224)))
            titles.append(f"baseline\n{label} | IoU={row['iou']:.2f}")

        for step in HEATMAP_STEPS:
            hm = _load_token_heatmap(sample_id, label, step)
            if hm is not None:
                imgs.append(hm)
                titles.append(f"step {step}\n{label}")

    out = VIS_DIR / f"03b_attention_heatmaps_{bucket}.png"
    make_grid(
        imgs,
        titles,
        ncols=4,
        cell_size=(224, 264),
        title_h=40,
        out_path=out,
    )
    display_saved(out)
    print(bucket, "examples:", len(chosen), "saved:", out)
```

## 6. Aggregate localization analytics

```python
loc_path = LOC_DIR / "attention_localization_rows.csv"
if loc_path.exists():
    import matplotlib.pyplot as plt
    loc_df = pd.read_csv(loc_path)
    if len(loc_df):
        by_step = loc_df.groupby("step", as_index=False).agg(
            iou_mean=("iou", "mean"),
            iou_median=("iou", "median"),
            iou_std=("iou", "std"),
            centroid_mean=("centroid_distance", "mean"),
            entropy_mean=("entropy", "mean"),
            n=("iou", "count"),
        )
        by_step.insert(0, "run", "sd15_800")
        display(by_step)

        comp = pd.concat([REFERENCE_150_LOC, by_step], ignore_index=True, sort=False)
        display(comp)

        fig = plt.figure(figsize=(6, 4))
        for run_name, grp in comp.groupby("run"):
            plt.plot(grp["step"], grp["iou_mean"], marker="o", label=run_name)
        plt.xlabel("Denoising step")
        plt.ylabel("Mean IoU")
        plt.title("Attention localization: 150 reference vs 800")
        plt.grid(True, alpha=0.3)
        plt.legend()
        out = VIS_DIR / "04_localization_iou_150_vs_800.png"
        fig.tight_layout()
        fig.savefig(out, dpi=160)
        plt.show()

        label_step = loc_df.groupby(["label", "step"], as_index=False).agg(n=("iou", "count"), iou_mean=("iou", "mean"), centroid_mean=("centroid_distance", "mean"))
        robust_labels = label_step[label_step["n"] >= 20].sort_values("iou_mean", ascending=False)
        print("Top labels by mean IoU, requiring n>=20 rows:")
        display(robust_labels.head(20))
        print("Low-IoU labels to inspect, requiring n>=20 rows:")
        display(robust_labels.tail(20))
```

## 8. Same-seed intervention comparison

```python
def parse_base_from_intervention_sample(sample_id: str) -> str | None:

    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"

map_rows = read_json(MAP_DIR / "intervention_rows.json", []) or []
sae_rows = read_json(SAE_STEER_DIR / "intervention_rows.json", []) or []

sae_by_base = {}
for row in sae_rows:
    base = parse_base_from_intervention_sample(row.get("sample_id", ""))
    if base and base not in sae_by_base:
        sae_by_base[base] = row

triples = []
seen = set()
for row in map_rows:
    base = parse_base_from_intervention_sample(row.get("sample_id", ""))
    if not base or base in seen:
        continue
    baseline_path = CAPTURE_DIR / "samples" / base / "image.png"
    map_path = MAP_DIR / "samples" / row["sample_id"] / "intervention.png"
    sae_row = sae_by_base.get(base)
    sae_path = SAE_STEER_DIR / "samples" / sae_row["sample_id"] / "steered.png" if sae_row else None
    if baseline_path.exists() and map_path.exists() and sae_path and sae_path.exists():
        triples.append((base, baseline_path, map_path, sae_path, row.get("target_word", "")))
        seen.add(base)
    if len(triples) >= 6:
        break

imgs, titles = [], []
for base, baseline_path, map_path, sae_path, target in triples:
    imgs.extend([open_image(baseline_path), open_image(map_path), open_image(sae_path)])
    short = base.replace("_seed_0000", "")
    titles.extend([f"baseline\n{short}", f"map x2\ntarget={target}", "SAE beta=2"])

out = VIS_DIR / "07_intervention_comparison_grid.png"
make_grid(imgs, titles, ncols=3, out_path=out)
display_saved(out)

if not triples:
    print("No complete baseline/map/SAE triples found yet. Run Stage 5 in the main notebook first.")
else:
    print("Displayed triples:", len(triples))
```

### new interventions

```python
import json
import math
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image
from IPython.display import display


def _read_json_safe(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _simple_l1(a_path: Path, b_path: Path, size=(192, 192)) -> float:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    return float(np.abs(y - x).mean())


def _collect_vis_rows(stage_dir: Path, *, image_name: str, kind: str) -> list[dict]:
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id", "")
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue

        baseline_path = CAPTURE_DIR / "samples" / base_id / "image.png"
        edited_path = stage_dir / "samples" / sample_id / image_name
        if not baseline_path.exists() or not edited_path.exists():
            continue

        out.append({
            "kind": kind,
            "sample_id": sample_id,
            "base_id": base_id,
            "label": row.get("target_word", ""),
            "strength": row.get("strength", np.nan),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "baseline_path": baseline_path,
            "edited_path": edited_path,
            "l1": _simple_l1(baseline_path, edited_path),
        })
    return out


candidate_sources = [
    (MAP_DIR, "intervention.png", "old_map"),
    (SAE_STEER_DIR, "steered.png", "old_sae"),
    (PROJECT_ROOT / "outputs/sd15_800/05b_map_reweight_strong_probe", "intervention.png", "strong_map"),
    (PROJECT_ROOT / "outputs/sd15_800/05b_block_scaling_probe", "ablation.png", "block_scaling"),
]

rows = []
for stage_dir, image_name, kind in candidate_sources:
    if stage_dir.exists():
        rows.extend(_collect_vis_rows(stage_dir, image_name=image_name, kind=kind))

sae_label_root = PROJECT_ROOT / "outputs/sd15_800/05b_sae_label_steering_probe"
if sae_label_root.exists():
    for label_dir in sorted(sae_label_root.iterdir()):
        if label_dir.is_dir():
            rows.extend(_collect_vis_rows(label_dir, image_name="steered.png", kind=f"sae_label/{label_dir.name}"))

df = pd.DataFrame(rows)
if not len(df):
    print("No intervention outputs found.")
else:
    display(
        df.groupby("kind")
        .agg(n=("l1", "count"), l1_mean=("l1", "mean"), l1_max=("l1", "max"))
        .sort_values("l1_mean", ascending=False)
    )

    top = df.sort_values("l1", ascending=False).head(12)

    imgs, titles = [], []
    for _, row in top.iterrows():
        imgs.append(open_image(Path(row["baseline_path"]), size=(224, 224)))
        imgs.append(open_image(Path(row["edited_path"]), size=(224, 224)))

        param = ""
        if not pd.isna(row.get("strength", np.nan)):
            param = f"x={row['strength']}"
        elif not pd.isna(row.get("beta", np.nan)):
            param = f"beta={row['beta']}"
        elif not pd.isna(row.get("scale", np.nan)):
            param = f"scale={row['scale']}"

        titles.append(f"baseline\n{row['base_id'][:32]}")
        titles.append(f"{row['kind']} {param}\nL1={row['l1']:.3f}")

    out = VIS_DIR / "08_intervention_most_changed_examples.png"
    make_grid(
        imgs,
        titles,
        ncols=4,
        cell_size=(224, 264),
        title_h=40,
        out_path=out,
    )
    display_saved(out)
    print("saved:", out)
```
