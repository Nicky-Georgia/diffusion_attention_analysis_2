```python
LINK_LARGE_DIRS = True
LARGE_DIRS = ["outputs", "annotations", "checkpoints", "runs"]


def copy_dir_contents(src: Path, dst: Path) -> None:
    dst.mkdir(parents=True, exist_ok=True)
    if any(src.iterdir()):
        if shutil.which("rsync"):
            subprocess.run(["rsync", "-a", f"{src}/", f"{dst}/"], check=True)
        else:
            shutil.copytree(src, dst, dirs_exist_ok=True)


def unique_backup_path(path: Path) -> Path:
    base = path.with_name(path.name + ".project_backup")
    if not base.exists():
        return base
    stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    return path.with_name(path.name + f".project_backup.{stamp}")


def ensure_storage_symlink(name: str) -> None:
    project_path = PROJECT_ROOT / name
    storage_path = STORAGE_PROJECT / name
    storage_path.mkdir(parents=True, exist_ok=True)

    if project_path.is_symlink():
        print(f"{name}: already symlink -> {project_path.resolve()}")
        return

    if project_path.exists():
        if not project_path.is_dir():
            raise RuntimeError(f"{project_path} exists and is not a directory")
        copy_dir_contents(project_path, storage_path)
        backup = unique_backup_path(project_path)
        project_path.rename(backup)
        project_path.symlink_to(storage_path, target_is_directory=True)
        print(f"{name}: moved project dir to {backup.name}; symlink -> {storage_path}")
    else:
        project_path.symlink_to(storage_path, target_is_directory=True)
        print(f"{name}: symlink -> {storage_path}")


if LINK_LARGE_DIRS:
    for dirname in LARGE_DIRS:
        ensure_storage_symlink(dirname)

print("\nResolved large dirs:")
for dirname in LARGE_DIRS:
    p = PROJECT_ROOT / dirname
    print(f"{dirname:12s}", "->", p.resolve() if p.exists() else "missing")
```

## 0.2 cleanup for previous sd15_150 artifacts

```python
CLEANUP_SD15_150 = True
CLEANUP_DRY_RUN = False
DELETE_SD15_150_PROMPT_FILES = False


def du_sh(path: Path) -> str:
    if not path.exists() and not path.is_symlink():
        return "missing"
    try:
        out = subprocess.check_output(["du", "-sh", str(path)], text=True, stderr=subprocess.DEVNULL)
        return out.split()[0]
    except Exception:
        return "?"


def cleanup_candidates_sd15_150() -> list[Path]:
    candidates = []
    for root in [PROJECT_ROOT, STORAGE_PROJECT]:
        candidates.extend([
            root / "outputs/sd15_150",
            root / "annotations/sd15_150",
            root / "runs/sd15_150",
            root / "checkpoints/sd15_150",
        ])
    if DELETE_SD15_150_PROMPT_FILES:
        candidates.extend([
            PROJECT_ROOT / "prompts/sd15_150_prompts.jsonl",
            PROJECT_ROOT / "prompts/sd15_150_intervention_prompts.jsonl",
        ])
    seen = set()
    unique = []
    for p in candidates:
        if not (p.exists() or p.is_symlink()):
            continue
        key = str(p.resolve()) if p.exists() else str(p)
        if key not in seen:
            seen.add(key)
            unique.append(p)
    return unique


def soft_cleanup_sd15_150(*, dry_run: bool = True) -> None:
    targets = cleanup_candidates_sd15_150()
    if not targets:
        print("No sd15_150 artifact directories/files found.")
        return

    print("Cleanup candidates for sd15_150 artifacts:")
    for p in targets:
        print(f"  {du_sh(p):>8s}  {p} -> {p.resolve() if p.exists() else 'missing'}")

    if dry_run:
        print("\nDRY_RUN=True: nothing deleted. Set CLEANUP_SD15_150=True and CLEANUP_DRY_RUN=False to delete.")
        return

    for p in targets:
        if p.is_symlink() or p.is_file():
            print("unlink", p)
            p.unlink(missing_ok=True)
        elif p.is_dir():
            print("rmtree", p)
            shutil.rmtree(p)
    print("Cleanup complete.")


soft_cleanup_sd15_150(dry_run=(not CLEANUP_SD15_150) or CLEANUP_DRY_RUN)
subprocess.run(["df", "-h", str(STORAGE)], check=False)
```

## Stage 2a — CLIPSeg pseudo-masks

```python
MASK_FORCE = False
MASK_MAX_SAMPLES = NUM_PROMPTS * NUM_SEEDS
MASK_THRESHOLD = 0.35
MASK_MIN_FRACTION = 0.005
MASK_FALLBACK_QUANTILE = 0.85


def generate_clipseg_masks(
    samples_dir: Path,
    output_dir: Path,
    *,
    max_samples: int | None = None,
    force: bool = False,
    threshold: float = 0.35,
    min_fraction: float = 0.005,
    fallback_quantile: float = 0.85,
) -> dict:
    import numpy as np
    import torch
    from PIL import Image
    from transformers import CLIPSegForImageSegmentation, CLIPSegProcessor

    if not samples_dir.exists():
        raise FileNotFoundError(samples_dir)

    output_dir.mkdir(parents=True, exist_ok=True)
    sample_dirs = sorted([p for p in samples_dir.iterdir() if p.is_dir()])
    if max_samples is not None:
        sample_dirs = sample_dirs[: int(max_samples)]

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model_id = "CIDAS/clipseg-rd64-refined"
    print("Loading CLIPSeg:", model_id, "device:", device)
    processor = CLIPSegProcessor.from_pretrained(model_id)
    model = CLIPSegForImageSegmentation.from_pretrained(model_id).to(device).eval()

    rows = []
    skipped = {"missing_image": 0, "missing_metadata": 0, "no_entities": 0}

    for idx, sample_dir in enumerate(sample_dirs, start=1):
        image_path = sample_dir / "image.png"
        metadata_path = sample_dir / "metadata.json"
        if not image_path.exists():
            skipped["missing_image"] += 1
            continue
        if not metadata_path.exists():
            skipped["missing_metadata"] += 1
            continue

        metadata = json.loads(metadata_path.read_text(encoding="utf-8"))
        prompt_record = metadata.get("prompt_record", {})
        labels = []
        for ent in prompt_record.get("entities", []) or []:
            if isinstance(ent, dict):
                label = str(ent.get("token") or ent.get("label") or ent.get("name") or "").strip().lower()
            else:
                label = str(ent).strip().lower()
            if label and label not in labels:
                labels.append(label)
        if not labels:
            skipped["no_entities"] += 1
            continue

        image = Image.open(image_path).convert("RGB")
        sample_out = output_dir / sample_dir.name
        sample_out.mkdir(parents=True, exist_ok=True)

        missing_labels = [label for label in labels if force or not (sample_out / f"{label.replace(' ', '_')}.png").exists()]
        if missing_labels:
            with torch.no_grad():
                inputs = processor(
                    text=missing_labels,
                    images=[image] * len(missing_labels),
                    return_tensors="pt",
                    padding=True,
                )
                inputs = {k: (v.to(device) if hasattr(v, "to") else v) for k, v in inputs.items()}
                logits = model(**inputs).logits
                if logits.ndim == 2:
                    logits = logits.unsqueeze(0)
                probs = torch.sigmoid(logits).unsqueeze(1)
                probs = torch.nn.functional.interpolate(
                    probs,
                    size=(image.height, image.width),
                    mode="bilinear",
                    align_corners=False,
                ).squeeze(1).detach().cpu().numpy()

            for label, prob in zip(missing_labels, probs):
                mask = prob >= threshold
                if float(mask.mean()) < min_fraction:
                    adaptive_t = float(np.quantile(prob, fallback_quantile))
                    mask = prob >= adaptive_t
                mask_path = sample_out / f"{label.replace(' ', '_')}.png"
                Image.fromarray((mask.astype(np.uint8) * 255)).save(mask_path)

        for label in labels:
            mask_path = sample_out / f"{label.replace(' ', '_')}.png"
            if mask_path.exists():
                arr = np.array(Image.open(mask_path).convert("L")) > 127
                rows.append({
                    "sample_id": sample_dir.name,
                    "prompt": metadata.get("prompt", ""),
                    "seed": metadata.get("seed"),
                    "label": label,
                    "mask_path": str(mask_path),
                    "mask_area_fraction": float(arr.mean()),
                })

        if idx % 25 == 0 or idx == len(sample_dirs):
            print(f"processed {idx}/{len(sample_dirs)} samples; masks={len(rows)}")

    manifest_columns = ["sample_id", "prompt", "seed", "label", "mask_path", "mask_area_fraction"]
    manifest_df = pd.DataFrame(rows, columns=manifest_columns)
    manifest_df.to_csv(output_dir / "mask_manifest.csv", index=False)
    (output_dir / "mask_manifest.json").write_text(json.dumps(rows, ensure_ascii=False, indent=2), encoding="utf-8")

    report = {
        "status": "ok" if rows else "skipped",
        "samples_seen": len(sample_dirs),
        "samples_with_masks": int(manifest_df["sample_id"].nunique()) if len(manifest_df) else 0,
        "masks": len(rows),
        "output_dir": str(output_dir),
        "threshold": threshold,
        "skipped": skipped,
    }
    (output_dir / "report.json").write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
    return report


if RUN_DRY_RUNS:
    print({
        "stage": "mask_generation_clipseg",
        "samples_dir": str(CAPTURE_DIR / "samples"),
        "output_dir": str(MASK_DIR),
        "max_samples": MASK_MAX_SAMPLES,
        "force": MASK_FORCE,
    })

if RUN_ACTUAL_STAGES:
    mask_report = generate_clipseg_masks(
        CAPTURE_DIR / "samples",
        MASK_DIR,
        max_samples=MASK_MAX_SAMPLES,
        force=MASK_FORCE,
        threshold=MASK_THRESHOLD,
        min_fraction=MASK_MIN_FRACTION,
        fallback_quantile=MASK_FALLBACK_QUANTILE,
    )
    print(mask_report)
```

```python
from pathlib import Path
import json
import math
import numpy as np
import pandas as pd
from PIL import Image
from IPython.display import display


def _read_json(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def _base_sample_id_from_intervention(sample_id: str) -> str | None:

    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }


def collect_intervention_diffs(stage_dir: Path, *, image_name: str, kind: str) -> pd.DataFrame:
    rows = _read_json(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue

        baseline_path = CAPTURE_DIR / "samples" / base_id / "image.png"
        intervention_path = stage_dir / "samples" / sample_id / image_name
        if not baseline_path.exists() or not intervention_path.exists():
            continue

        meta = {
            "kind": kind,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "target_word": row.get("target_word", ""),
            "strength": row.get("strength", np.nan),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "image_path": str(intervention_path),
            "baseline_path": str(baseline_path),
        }
        meta.update(_image_diff_metrics(baseline_path, intervention_path))
        out.append(meta)

    return pd.DataFrame(out)


diff_tables = []

if MAP_DIR.exists():
    diff_tables.append(
        collect_intervention_diffs(MAP_DIR, image_name="intervention.png", kind="old_map_x2_step24")
    )

if SAE_STEER_DIR.exists():
    diff_tables.append(
        collect_intervention_diffs(SAE_STEER_DIR, image_name="steered.png", kind="old_sae_beta2_step24")
    )

EXTRA_INTERVENTION_DIRS = [
    (PROJECT_ROOT / "outputs/sd15_800/05b_map_reweight_strong_probe", "intervention.png", "strong_map_probe"),
    (PROJECT_ROOT / "outputs/sd15_800/05b_block_scaling_probe", "ablation.png", "block_scaling_probe"),
]

for stage_dir, image_name, kind in EXTRA_INTERVENTION_DIRS:
    if stage_dir.exists():
        diff_tables.append(
            collect_intervention_diffs(stage_dir, image_name=image_name, kind=kind)
        )

SAE_LABEL_PROBE_ROOT = PROJECT_ROOT / "outputs/sd15_800/05b_sae_label_steering_probe"
if SAE_LABEL_PROBE_ROOT.exists():
    for label_dir in sorted(SAE_LABEL_PROBE_ROOT.iterdir()):
        if label_dir.is_dir():
            diff_tables.append(
                collect_intervention_diffs(label_dir, image_name="steered.png", kind=f"sae_label_probe/{label_dir.name}")
            )

all_diffs = pd.concat([x for x in diff_tables if len(x)], ignore_index=True) if diff_tables else pd.DataFrame()

if len(all_diffs):
    print("Intervention image-difference summary:")
    display(
        all_diffs.groupby("kind")
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_median=("l1_mean", "median"),
            rmse_mean=("rmse", "mean"),
            changed_px_005=("changed_px_005", "mean"),
            changed_px_010=("changed_px_010", "mean"),
            changed_px_020=("changed_px_020", "mean"),
        )
        .sort_values("l1_mean", ascending=False)
    )

    print("Most changed examples:")
    display(
        all_diffs.sort_values("l1_mean", ascending=False)
        .head(20)[["kind", "sample_id", "target_word", "strength", "beta", "scale", "l1_mean", "rmse", "changed_px_010"]]
    )

    print("Least changed examples:")
    display(
        all_diffs.sort_values("l1_mean", ascending=True)
        .head(20)[["kind", "sample_id", "target_word", "strength", "beta", "scale", "l1_mean", "rmse", "changed_px_010"]]
    )
else:
    print("No intervention images found for diff diagnostics yet.")
```

```python
import json
from pathlib import Path
import pandas as pd
from IPython.display import display


def _yaml_list(xs):
    return "[" + ",".join(str(x) for x in xs) + "]"


def _target_word_from_record(row: dict) -> str:
    if row.get("target_word"):
        return str(row["target_word"]).lower()
    entities = row.get("entities") or []
    if entities:
        return str(entities[0].get("token") or entities[0].get("label")).lower()
    return ""


def _read_jsonl(path: Path) -> list[dict]:
    rows = []
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


PROBE_PROMPTS_PER_LABEL = 4
FEATURES_PER_LABEL = 5

PREFERRED_PROBE_LABELS = [
    "banana", "bus", "boat", "hat", "apple", "bear",
    "dog", "cat", "book", "car", "cup", "flower",
    "bicycle", "vase", "umbrella", "phone",
]

concept_csv = CONCEPT_DIR / "concept_dictionary.csv"
assert concept_csv.exists(), f"Missing concept dictionary: {concept_csv}"

concept_df = pd.read_csv(concept_csv)
concept_df["label"] = concept_df["label"].astype(str).str.lower()
concept_df["feature_id"] = pd.to_numeric(concept_df["feature_id"], errors="coerce").astype("Int64")
concept_df["score"] = pd.to_numeric(concept_df.get("score", 0.0), errors="coerce").fillna(0.0)

intervention_records_all = _read_jsonl(INTERVENTION_PROMPTS_800)
for row in intervention_records_all:
    row["target_word"] = _target_word_from_record(row)

target_counts = pd.Series([r["target_word"] for r in intervention_records_all]).value_counts().to_dict()
concept_labels = set(concept_df["label"].dropna().unique().tolist())

TARGET_LABELS_FOR_PROBE = [
    label for label in PREFERRED_PROBE_LABELS
    if target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL and label in concept_labels
]

print("Probe labels:", TARGET_LABELS_FOR_PROBE)

PROBE_PROMPT_ROOT = PROJECT_ROOT / "prompts/sd15_800_stage5_probes"
PROBE_PROMPT_ROOT.mkdir(parents=True, exist_ok=True)

probe_records = []
probe_label_prompt_paths = {}
label_to_concept_ids = {}

for label in TARGET_LABELS_FOR_PROBE:
    label_records = [r for r in intervention_records_all if r["target_word"] == label][:PROBE_PROMPTS_PER_LABEL]
    if len(label_records) < PROBE_PROMPTS_PER_LABEL:
        continue

    path = PROBE_PROMPT_ROOT / f"sd15_800_probe_{label}.jsonl"
    with path.open("w", encoding="utf-8") as f:
        for r in label_records:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")
    probe_label_prompt_paths[label] = path
    probe_records.extend(label_records)

    top_features = (
        concept_df[concept_df["label"] == label]
        .sort_values("score", ascending=False)
        .dropna(subset=["feature_id"])
        .head(FEATURES_PER_LABEL)["feature_id"]
        .astype(int)
        .tolist()
    )
    label_to_concept_ids[label] = top_features

PROBE_PROMPTS_ALL = PROBE_PROMPT_ROOT / "sd15_800_probe_all_labels.jsonl"
with PROBE_PROMPTS_ALL.open("w", encoding="utf-8") as f:
    for r in probe_records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

probe_summary = []
for label in TARGET_LABELS_FOR_PROBE:
    probe_summary.append({
        "label": label,
        "prompt_file": str(probe_label_prompt_paths.get(label)),
        "n_prompts": sum(1 for r in probe_records if r["target_word"] == label),
        "concept_ids": label_to_concept_ids.get(label),
        "top_score": float(
            concept_df[concept_df["label"] == label].sort_values("score", ascending=False)["score"].head(1).iloc[0]
        ),
    })

display(pd.DataFrame(probe_summary))
print("Combined probe prompts:", PROBE_PROMPTS_ALL)
print("Total probe prompts:", len(probe_records))
```
