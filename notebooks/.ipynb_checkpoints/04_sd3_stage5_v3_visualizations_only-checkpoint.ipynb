## 1. Stage 5 v3 artifact inventory

```python
def _read_json_safe(path: Path, default=None):
    if not path.exists():
        return default
    try:
        return json.loads(path.read_text(encoding="utf-8"))
    except Exception as e:
        print("[warn] cannot read", path, e)
        return default

def _report_row(stage_dir: Path, kind: str, **extra):
    report = _read_json_safe(stage_dir / "report.json", {}) or {}
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    samples_dir = stage_dir / "samples"
    n_png = len(list(samples_dir.glob("*/ablation.png"))) + len(list(samples_dir.glob("*/steered.png"))) if samples_dir.exists() else 0
    return {
        "kind": kind,
        "stage_dir": str(stage_dir),
        "exists": stage_dir.exists(),
        "status": report.get("status"),
        "rows_in_report": report.get("rows") or report.get("n_rows") or report.get("num_rows"),
        "intervention_rows": len(rows),
        "png_outputs": n_png,
        **extra,
    }

inventory = []

if BLOCK_V3_ROOT.exists():
    for d in sorted([p for p in BLOCK_V3_ROOT.iterdir() if p.is_dir()]):
        name = d.name
        module, window = None, None
        if "__" in name:
            module_part, window = name.rsplit("__", 1)
            module = module_part.replace("transformer_blocks_", "transformer_blocks.")
        inventory.append(_report_row(d, "block_v3", module=module, window=window))

for root, kind in [(SAE_GLOBAL_V3_ROOT, "sae_global_v3"), (SAE_ACTIVE_V3_ROOT, "sae_active_tokens_v3")]:
    if root.exists():
        for label_dir in sorted([p for p in root.iterdir() if p.is_dir()]):
            for mode_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
                for window_dir in sorted([p for p in mode_dir.iterdir() if p.is_dir()]):
                    inventory.append(_report_row(
                        window_dir,
                        kind,
                        label=label_dir.name,
                        mode=mode_dir.name,
                        window=window_dir.name,
                    ))

inventory_df = pd.DataFrame(inventory)
print("Stage 5 v3 sub-runs:", len(inventory_df))
if len(inventory_df):
    display(inventory_df.sort_values(["kind", "label", "module", "window"], na_position="last"))
    inventory_df.to_csv(VIS_V3_DIR / "stage5_v3_inventory.csv", index=False)
else:
    print("No Stage 5 v3 sub-runs found.")
```

## 3. Collect or load same-seed image diffs

```python
from PIL import Image, ImageChops

def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"

def _find_baseline_image(base_id: str) -> Path | None:
    candidates = [
        SPATIAL_CAPTURE_DIR / "samples" / base_id / "image.png",
        ORIGINAL_CAPTURE_DIR / "samples" / base_id / "image.png",
    ]
    for p in candidates:
        if p.exists():
            return p
    return None

def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
        "changed_px_030": float((diff_gray > 0.30).mean()),
    }

def _first_requested_module(row):
    mods = row.get("requested_modules")
    if isinstance(mods, list) and mods:
        return mods[0]
    return row.get("module", "")

def collect_diffs(stage_dir: Path, *, image_name: str, kind: str, label: str | None = None, module: str | None = None, window: str | None = None, mode: str | None = None) -> pd.DataFrame:
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = _find_baseline_image(base_id)
        edited_path = stage_dir / "samples" / sample_id / image_name
        if baseline_path is None or not edited_path.exists():
            continue
        meta = {
            "kind": kind,
            "label": label or row.get("target_word", ""),
            "module": module or _first_requested_module(row),
            "window": window,
            "mode": mode,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "baseline_path": str(baseline_path),
            "edited_path": str(edited_path),
        }
        meta.update(_image_diff_metrics(baseline_path, edited_path))
        out.append(meta)
    return pd.DataFrame(out)

FORCE_RECOMPUTE_DIFFS = False

if STAGE5_V3_DIFFS_CSV.exists() and not FORCE_RECOMPUTE_DIFFS:
    all_diffs_v3 = pd.read_csv(STAGE5_V3_DIFFS_CSV)
    print("Loaded existing diffs:", STAGE5_V3_DIFFS_CSV)
else:
    diff_tables = []
    if BLOCK_V3_ROOT.exists():
        for d in sorted([p for p in BLOCK_V3_ROOT.iterdir() if p.is_dir()]):
            name = d.name
            if "__" in name:
                module_part, window = name.rsplit("__", 1)
                module = module_part.replace("transformer_blocks_", "transformer_blocks.")
            else:
                module, window = None, None
            diff_tables.append(collect_diffs(d, image_name="ablation.png", kind="block_v3", module=module, window=window))

    for root, kind in [(SAE_GLOBAL_V3_ROOT, "sae_global_v3"), (SAE_ACTIVE_V3_ROOT, "sae_active_tokens_v3")]:
        if root.exists():
            for label_dir in sorted([p for p in root.iterdir() if p.is_dir()]):
                for mode_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
                    for window_dir in sorted([p for p in mode_dir.iterdir() if p.is_dir()]):
                        diff_tables.append(collect_diffs(
                            window_dir,
                            image_name="steered.png",
                            kind=kind,
                            label=label_dir.name,
                            mode=mode_dir.name,
                            window=window_dir.name,
                        ))

    non_empty = [d for d in diff_tables if len(d)]
    all_diffs_v3 = pd.concat(non_empty, ignore_index=True) if non_empty else pd.DataFrame()
    if len(all_diffs_v3):
        STAGE5_V3_DIFFS_CSV.parent.mkdir(parents=True, exist_ok=True)
        all_diffs_v3.to_csv(STAGE5_V3_DIFFS_CSV, index=False)
        print("Saved diffs:", STAGE5_V3_DIFFS_CSV)

if len(all_diffs_v3):
    all_diffs_v3["artifact_flag"] = (
        (pd.to_numeric(all_diffs_v3["l1_mean"], errors="coerce") > 0.22)
        | (pd.to_numeric(all_diffs_v3["changed_px_020"], errors="coerce") > 0.55)
        | (pd.to_numeric(all_diffs_v3["changed_px_030"], errors="coerce") > 0.35)
    )
    all_diffs_v3["moderate_change"] = (
        (pd.to_numeric(all_diffs_v3["l1_mean"], errors="coerce") >= 0.015)
        & (pd.to_numeric(all_diffs_v3["l1_mean"], errors="coerce") <= 0.18)
        & (~all_diffs_v3["artifact_flag"])
    )

    all_diffs_v3.to_csv(VIS_V3_DIR / "stage5_v3_image_diffs_enriched.csv", index=False)

print("Diff rows:", len(all_diffs_v3))
if len(all_diffs_v3):
    display(all_diffs_v3.head())
```

## 4. Overall Stage 5 v3 summary

```python
def _agg(df, by):
    if not len(df):
        return pd.DataFrame()
    return (
        df.groupby(by, dropna=False)
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_median=("l1_mean", "median"),
            l1_max=("l1_mean", "max"),
            changed_px_010=("changed_px_010", "mean"),
            changed_px_020=("changed_px_020", "mean"),
            artifact_rate=("artifact_flag", "mean"),
            moderate_rate=("moderate_change", "mean"),
        )
        .reset_index()
        .sort_values(["l1_mean", "changed_px_020"], ascending=False)
    )

if len(all_diffs_v3):
    overall = _agg(all_diffs_v3, ["kind"])
    display(overall)
    overall.to_csv(VIS_V3_DIR / "stage5_v3_overall_summary.csv", index=False)

    by_window = _agg(all_diffs_v3, ["kind", "window"])
    display(by_window)
    by_window.to_csv(VIS_V3_DIR / "stage5_v3_by_window_summary.csv", index=False)
```

## 9. Unicode-safe gallery helpers

```python
from PIL import ImageDraw, ImageFont

def _load_font(size=13):
    for path in [
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
        "/usr/local/share/fonts/DejaVuSans.ttf",
    ]:
        if Path(path).exists():
            return ImageFont.truetype(path, size=size)
    return ImageFont.load_default()

FONT = _load_font(13)

def open_image(path: Path, size=(224, 224)):
    return Image.open(path).convert("RGB").resize(size)

def diff_image(a_path: Path, b_path: Path, size=(224, 224), amplify: float = 3.0):
    a = open_image(a_path, size)
    b = open_image(b_path, size)
    d = ImageChops.difference(a, b)
    arr = np.asarray(d).astype(np.float32)
    arr = np.clip(arr * amplify, 0, 255).astype(np.uint8)
    return Image.fromarray(arr)

def make_grid(images, titles, ncols=3, cell_size=(224, 264), title_h=40, out_path=None):
    if not images:
        print("No images for grid", out_path)
        return None
    nrows = math.ceil(len(images) / ncols)
    W, H = cell_size
    grid = Image.new("RGB", (ncols * W, nrows * H), "white")
    draw = ImageDraw.Draw(grid)
    for idx, img in enumerate(images):
        r, c = divmod(idx, ncols)
        x, y = c * W, r * H
        grid.paste(img, (x, y + title_h))
        title = str(titles[idx])
        draw.multiline_text((x + 6, y + 5), title[:110], fill=(0, 0, 0), font=FONT, spacing=2)
    if out_path:
        grid.save(out_path)
        display(grid)
        print("Saved:", out_path)
    return grid

def _param_title(row):
    if "beta" in row.index and not pd.isna(row.get("beta", np.nan)):
        return f"beta={float(row['beta']):g}"
    if "scale" in row.index and not pd.isna(row.get("scale", np.nan)):
        return f"scale={float(row['scale']):g}"
    return ""

def make_triplet_gallery(rows: pd.DataFrame, out_path: Path, *, title_prefix: str, max_rows: int = 8, sort_note: str = ""):
    imgs, titles = [], []
    rows = rows.head(max_rows)
    for _, row in rows.iterrows():
        base = Path(row["baseline_path"])
        edit = Path(row["edited_path"])
        if not base.exists() or not edit.exists():
            continue
        imgs.extend([open_image(base), open_image(edit), diff_image(base, edit)])
        label = str(row.get("label", "") or "")
        window = str(row.get("window", "") or "")
        mode = str(row.get("mode", "") or "")
        module = str(row.get("module", "") or "")
        extra = " / ".join([x for x in [label, mode, window, module] if x and x != "nan"])
        titles.extend([
            f"baseline\n{str(row['base_id'])[:26]}",
            f"{title_prefix} {_param_title(row)}\n{extra[:34]}\nL1={row['l1_mean']:.3f}",
            f"diff {sort_note}\npx>.10={row['changed_px_010']:.2f}\npx>.20={row['changed_px_020']:.2f}",
        ])
    return make_grid(imgs, titles, ncols=3, out_path=out_path)

def best_per_group(df: pd.DataFrame, group_cols, *, sort_col="l1_mean", min_l1=None, max_l1=None):
    x = df.copy()
    if min_l1 is not None:
        x = x[x["l1_mean"] >= min_l1]
    if max_l1 is not None:
        x = x[x["l1_mean"] <= max_l1]
    if not len(x):
        return x
    return (
        x.sort_values(sort_col, ascending=False)
        .groupby(group_cols, as_index=False, dropna=False)
        .head(1)
        .sort_values(sort_col, ascending=False)
    )
```

## 11. Optional CLIPSeg semantic check

```python
RUN_CLIPSEG_STAGE5_V3_EVAL = False
CLIPSEG_MAX_ROWS = 80


if RUN_CLIPSEG_STAGE5_V3_EVAL and len(all_diffs_v3):
    import torch
    from transformers import CLIPSegProcessor, CLIPSegForImageSegmentation

    device = "cuda" if torch.cuda.is_available() else "cpu"
    processor = CLIPSegProcessor.from_pretrained("CIDAS/clipseg-rd64-refined")
    model = CLIPSegForImageSegmentation.from_pretrained("CIDAS/clipseg-rd64-refined").to(device).eval()

    def clipseg_score(image_path: Path, label: str) -> float:
        image = Image.open(image_path).convert("RGB")
        inputs = processor(text=[label], images=[image], padding=True, return_tensors="pt").to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
            prob = torch.sigmoid(logits).float().mean().item()
        return float(prob)

    eval_rows = []
    rows = all_diffs_v3[all_diffs_v3["kind"].astype(str).str.startswith("sae")].copy()
    rows = rows[(~rows["artifact_flag"]) & (rows["label"].astype(str).str.len() > 0)]
    rows = rows.sort_values("l1_mean", ascending=False).head(CLIPSEG_MAX_ROWS)

    for _, row in rows.iterrows():
        label = str(row["label"])
        before = clipseg_score(Path(row["baseline_path"]), label)
        after = clipseg_score(Path(row["edited_path"]), label)
        d = row.to_dict()
        d.update({"clipseg_before": before, "clipseg_after": after, "clipseg_delta": after - before})
        eval_rows.append(d)

    clipseg_v3 = pd.DataFrame(eval_rows)
    out_csv = VIS_V3_DIR / "stage5_v3_clipseg_eval.csv"
    clipseg_v3.to_csv(out_csv, index=False)
    print("Saved:", out_csv)

    clipseg_summary = (
        clipseg_v3.groupby(["kind", "label", "mode", "window", "beta"], dropna=False)
        .agg(
            n=("clipseg_delta", "count"),
            delta_mean=("clipseg_delta", "mean"),
            delta_median=("clipseg_delta", "median"),
            before=("clipseg_before", "mean"),
            after=("clipseg_after", "mean"),
            l1_mean=("l1_mean", "mean"),
            artifact_rate=("artifact_flag", "mean"),
        )
        .reset_index()
        .sort_values("delta_mean", ascending=False)
    )
    display(clipseg_summary.head(60))
    clipseg_summary.to_csv(VIS_V3_DIR / "stage5_v3_clipseg_summary.csv", index=False)
else:
    existing = VIS_V3_DIR / "stage5_v3_clipseg_eval.csv"
    if existing.exists():
        clipseg_v3 = pd.read_csv(existing)
        print("Loaded existing CLIPSeg eval:", existing)
        display(clipseg_v3.head())
```
