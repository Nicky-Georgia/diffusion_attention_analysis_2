## 0.3 Build deterministic prompt files

```python
import json
from pathlib import Path

REWRITE_PROMPT_FILES = True

OBJECTS = [
    "dog", "cat", "car", "bicycle", "cup", "book", "vase", "flower", "person", "umbrella",
    "tree", "bench", "chair", "table", "bird", "horse", "boat", "laptop", "backpack", "bottle",
    "apple", "banana", "clock", "train", "bus", "sofa", "bed", "window", "skateboard", "airplane",
]
RELATIONS = [
    "next to", "beside", "near", "in front of", "behind", "to the left of", "to the right of",
    "under", "above", "on",
]
SCENES = [
    "in a city park", "on a quiet street", "inside a bright room", "on a wooden floor", "near a small garden",
    "on green grass", "beside a stone wall", "in soft daylight", "on a clean background", "near a window",
    "at the beach", "in a cozy kitchen", "in a modern living room", "under a cloudy sky", "on a market square",
]
STYLES = ["realistic photo", "studio photo", "documentary photo", "cinematic photo", "high-detail digital image"]


def article(label: str) -> str:
    return "an" if label[0].lower() in "aeiou" else "a"


def make_prompt_rows(n: int, prefix: str):
    rows = []
    for i in range(n):
        obj_a = OBJECTS[i % len(OBJECTS)]
        obj_b = OBJECTS[(i * 7 + 5) % len(OBJECTS)]
        if obj_b == obj_a:
            obj_b = OBJECTS[(i + 1) % len(OBJECTS)]
        relation = RELATIONS[(i // len(OBJECTS) + i) % len(RELATIONS)]
        scene = SCENES[(i * 3 + 2) % len(SCENES)]
        style = STYLES[i % len(STYLES)]
        prompt = f"{style} of {article(obj_a)} {obj_a} {relation} {article(obj_b)} {obj_b} {scene}"
        rows.append({
            "id": f"{prefix}_{i:04d}",
            "prompt": prompt,
            "entities": [
                {"label": obj_a, "token": obj_a},
                {"label": obj_b, "token": obj_b},
            ],
        })
    return rows


def write_jsonl(path: Path, rows):
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8") as f:
        for row in rows:
            f.write(json.dumps(row, ensure_ascii=False) + "\n")

if REWRITE_PROMPT_FILES or not PROMPTS_150.exists():
    rows_150 = make_prompt_rows(SAMPLE_TARGET, "sd3m150")
    write_jsonl(PROMPTS_150, rows_150)
else:
    rows_150 = [json.loads(line) for line in PROMPTS_150.read_text(encoding="utf-8").splitlines() if line.strip()]

if REWRITE_PROMPT_FILES or not ABLATION_PROMPTS_50.exists():
    write_jsonl(ABLATION_PROMPTS_50, rows_150[:ABLATION_PROMPT_TARGET])

print("Main prompt rows:", sum(1 for _ in PROMPTS_150.open("r", encoding="utf-8")))
print("Ablation prompt rows:", sum(1 for _ in ABLATION_PROMPTS_50.open("r", encoding="utf-8")))
print("Example:", PROMPTS_150.read_text(encoding="utf-8").splitlines()[0])
```

### pseudo-masks for SD3 concept dictionarypseudo-masks for SD3 concept dictionary

```python
import json
from pathlib import Path

MASK_FORCE = True
MASK_MAX_SAMPLES = SAMPLE_TARGET * NUM_SEEDS
MASK_THRESHOLD = 0.35
MASK_MIN_FRACTION = 0.005
MASK_FALLBACK_QUANTILE = 0.85


def _label_to_filename(label: str) -> str:
    return str(label).strip().lower().replace(" ", "_").replace("/", "_")


def generate_clipseg_masks_sd3(
    samples_dir: Path,
    output_dir: Path,
    *,
    max_samples: int | None = None,
    force: bool = False,
    threshold: float = 0.35,
    min_fraction: float = 0.005,
    fallback_quantile: float = 0.85,
) -> dict:
    import numpy as np
    import torch
    from PIL import Image
    from transformers import CLIPSegForImageSegmentation, CLIPSegProcessor

    if not samples_dir.exists():
        raise FileNotFoundError(samples_dir)

    output_dir.mkdir(parents=True, exist_ok=True)
    sample_dirs = sorted([p for p in samples_dir.iterdir() if p.is_dir()])
    if max_samples is not None:
        sample_dirs = sample_dirs[: int(max_samples)]

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model_id = "CIDAS/clipseg-rd64-refined"
    print("Loading CLIPSeg:", model_id, "device:", device)
    processor = CLIPSegProcessor.from_pretrained(model_id)
    model = CLIPSegForImageSegmentation.from_pretrained(model_id).to(device).eval()

    rows = []
    skipped = {"missing_image": 0, "missing_metadata": 0, "no_entities": 0}

    for idx, sample_dir in enumerate(sample_dirs, start=1):
        image_path = sample_dir / "image.png"
        metadata_path = sample_dir / "metadata.json"
        if not image_path.exists():
            skipped["missing_image"] += 1
            continue
        if not metadata_path.exists():
            skipped["missing_metadata"] += 1
            continue

        metadata = json.loads(metadata_path.read_text(encoding="utf-8"))
        prompt_record = metadata.get("prompt_record", {})
        labels = []
        for ent in prompt_record.get("entities", []) or []:
            if isinstance(ent, dict):
                label = str(ent.get("token") or ent.get("label") or ent.get("name") or "").strip().lower()
            else:
                label = str(ent).strip().lower()
            if label and label not in labels:
                labels.append(label)
        if not labels:
            skipped["no_entities"] += 1
            continue

        image = Image.open(image_path).convert("RGB")
        sample_out = output_dir / sample_dir.name
        sample_out.mkdir(parents=True, exist_ok=True)

        missing_labels = [label for label in labels if force or not (sample_out / f"{_label_to_filename(label)}.png").exists()]
        if missing_labels:
            with torch.no_grad():
                inputs = processor(
                    text=missing_labels,
                    images=[image] * len(missing_labels),
                    return_tensors="pt",
                    padding=True,
                )
                inputs = {k: (v.to(device) if hasattr(v, "to") else v) for k, v in inputs.items()}
                logits = model(**inputs).logits
                if logits.ndim == 2:
                    logits = logits.unsqueeze(0)
                probs = torch.sigmoid(logits).unsqueeze(1)
                probs = torch.nn.functional.interpolate(
                    probs,
                    size=(image.height, image.width),
                    mode="bilinear",
                    align_corners=False,
                ).squeeze(1).detach().cpu().numpy()

            for label, prob in zip(missing_labels, probs):
                mask = prob >= threshold
                if float(mask.mean()) < min_fraction:
                    adaptive_t = float(np.quantile(prob, fallback_quantile))
                    mask = prob >= adaptive_t
                mask_path = sample_out / f"{_label_to_filename(label)}.png"
                Image.fromarray((mask.astype(np.uint8) * 255)).save(mask_path)

        for label in labels:
            mask_path = sample_out / f"{_label_to_filename(label)}.png"
            if mask_path.exists():
                arr = np.array(Image.open(mask_path).convert("L")) > 127
                rows.append({
                    "sample_id": sample_dir.name,
                    "prompt": metadata.get("prompt", ""),
                    "seed": metadata.get("seed"),
                    "label": label,
                    "mask_path": str(mask_path),
                    "mask_area_fraction": float(arr.mean()),
                })

        if idx % 25 == 0 or idx == len(sample_dirs):
            print(f"processed {idx}/{len(sample_dirs)} samples; masks={len(rows)}")

    manifest_columns = ["sample_id", "prompt", "seed", "label", "mask_path", "mask_area_fraction"]
    manifest_df = pd.DataFrame(rows, columns=manifest_columns)
    manifest_df.to_csv(output_dir / "mask_manifest.csv", index=False)
    (output_dir / "mask_manifest.json").write_text(json.dumps(rows, ensure_ascii=False, indent=2), encoding="utf-8")

    report = {
        "status": "ok" if rows else "skipped",
        "samples_seen": len(sample_dirs),
        "samples_with_masks": int(manifest_df["sample_id"].nunique()) if len(manifest_df) else 0,
        "masks": len(rows),
        "output_dir": str(output_dir),
        "threshold": threshold,
        "fallback_quantile": fallback_quantile,
        "skipped": skipped,
    }
    (output_dir / "report.json").write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
    return report


mask_report = generate_clipseg_masks_sd3(
    CAPTURE_DIR / "samples",
    MASK_DIR,
    max_samples=MASK_MAX_SAMPLES,
    force=MASK_FORCE,
    threshold=MASK_THRESHOLD,
    min_fraction=MASK_MIN_FRACTION,
    fallback_quantile=MASK_FALLBACK_QUANTILE,
)
print(mask_report)

mask_manifest_path = MASK_DIR / "mask_manifest.csv"
assert mask_manifest_path.exists(), mask_manifest_path
mask_df = pd.read_csv(mask_manifest_path)
print("mask rows:", len(mask_df))
print("samples with masks:", mask_df["sample_id"].nunique())
print("masks per sample:", len(mask_df) / max(mask_df["sample_id"].nunique(), 1))
display(mask_df.groupby("label").agg(samples=("sample_id", "nunique"), mean_area=("mask_area_fraction", "mean")).sort_values("samples", ascending=False).head(30))
```

### prompt subsets and SD3-specific intervention configs

```python
import re


def _yaml_list(xs):
    return "[" + ",".join(str(x) for x in xs) + "]"


def _safe_file_part(text: str) -> str:
    text = str(text).strip().lower()
    text = re.sub(r"[^a-z0-9_-]+", "_", text)
    text = re.sub(r"_+", "_", text).strip("_")
    return text or "label"


def _target_word_from_record(row: dict) -> str:
    if row.get("target_word"):
        return str(row["target_word"]).lower()
    for key in ("target", "label"):
        if row.get(key):
            return str(row[key]).lower()
    entities = row.get("entities") or []
    if entities:
        ent = entities[0]
        if isinstance(ent, dict):
            return str(ent.get("token") or ent.get("label") or ent.get("name") or "").lower()
        return str(ent).lower()
    return ""


def _read_jsonl(path: Path) -> list[dict]:
    rows = []
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


PROBE_PROMPTS_PER_LABEL = 3
FEATURES_PER_LABEL = 5
MAX_STAGE5_LABELS = 6

PREFERRED_PROBE_LABELS = [
    "dog", "cat", "car", "bicycle", "cup", "book", "vase", "flower", "person", "umbrella",
    "tree", "bench", "chair", "table", "bird", "horse", "boat", "laptop", "backpack", "bottle",
    "apple", "banana", "clock", "train", "bus", "sofa", "bed", "window", "skateboard", "airplane",
]

concept_csv = CONCEPT_DIR / "concept_dictionary.csv"
assert concept_csv.exists(), concept_csv
concept_df = pd.read_csv(concept_csv)
concept_df["label"] = concept_df["label"].astype(str).str.lower()
concept_df["feature_id"] = pd.to_numeric(concept_df["feature_id"], errors="coerce").astype("Int64")
concept_df["score"] = pd.to_numeric(concept_df.get("score", 0.0), errors="coerce").fillna(0.0)

prompt_records_all = _read_jsonl(PROMPTS_150)
for row in prompt_records_all:
    row["target_word"] = _target_word_from_record(row)

target_counts = pd.Series([r["target_word"] for r in prompt_records_all if r.get("target_word")]).value_counts().to_dict()
concept_labels = set(concept_df["label"].dropna().unique().tolist())

candidate_labels = [
    label for label in PREFERRED_PROBE_LABELS
    if target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL and label in concept_labels
]

# Fallback: if preferred labels are sparse, use the strongest labels that also occur in the prompt file.
if len(candidate_labels) < MAX_STAGE5_LABELS:
    top_concept_labels = (
        concept_df.groupby("label", as_index=False)
        .agg(top_score=("score", "max"), n_features=("feature_id", "count"))
        .sort_values("top_score", ascending=False)["label"]
        .tolist()
    )
    for label in top_concept_labels:
        if label not in candidate_labels and target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL:
            candidate_labels.append(label)
        if len(candidate_labels) >= MAX_STAGE5_LABELS:
            break

TARGET_LABELS_FOR_PROBE = candidate_labels[:MAX_STAGE5_LABELS]
assert TARGET_LABELS_FOR_PROBE, "No labels are available for Stage 5 probes. Check concept_df labels and prompt entities."
print("Probe labels:", TARGET_LABELS_FOR_PROBE)

PROBE_PROMPT_ROOT.mkdir(parents=True, exist_ok=True)
probe_records = []
probe_label_prompt_paths = {}
label_to_concept_ids = {}

for label in TARGET_LABELS_FOR_PROBE:
    label_records = [dict(r) for r in prompt_records_all if r.get("target_word") == label][:PROBE_PROMPTS_PER_LABEL]
    if len(label_records) < PROBE_PROMPTS_PER_LABEL:
        print(f"[skip] {label}: only {len(label_records)} prompt records")
        continue

    for r in label_records:
        r["target_word"] = label

    prompt_path = PROBE_PROMPT_ROOT / f"sd3_probe_{_safe_file_part(label)}.jsonl"
    with prompt_path.open("w", encoding="utf-8") as f:
        for r in label_records:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")
    probe_label_prompt_paths[label] = prompt_path
    probe_records.extend(label_records)

    top_features = (
        concept_df[concept_df["label"] == label]
        .sort_values("score", ascending=False)
        .dropna(subset=["feature_id"])
        .head(FEATURES_PER_LABEL)["feature_id"]
        .astype(int)
        .tolist()
    )
    label_to_concept_ids[label] = top_features

PROBE_PROMPTS_ALL = PROBE_PROMPT_ROOT / "sd3_probe_all_labels.jsonl"
with PROBE_PROMPTS_ALL.open("w", encoding="utf-8") as f:
    for r in probe_records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

BLOCK_PROBE_PROMPTS = PROBE_PROMPT_ROOT / "sd3_block_probe.jsonl"
with BLOCK_PROBE_PROMPTS.open("w", encoding="utf-8") as f:
    for r in probe_records[: min(12, len(probe_records))]:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

probe_summary = []
for label in TARGET_LABELS_FOR_PROBE:
    probe_summary.append({
        "label": label,
        "prompt_file": str(probe_label_prompt_paths.get(label)),
        "n_prompts": sum(1 for r in probe_records if r.get("target_word") == label),
        "concept_ids": label_to_concept_ids.get(label),
        "top_score": float(concept_df[concept_df["label"] == label].sort_values("score", ascending=False)["score"].head(1).iloc[0]),
    })

display(pd.DataFrame(probe_summary))
print("Combined probe prompts:", PROBE_PROMPTS_ALL)
print("Block probe prompts:", BLOCK_PROBE_PROMPTS)
print("Total SAE probe prompts:", len(probe_records))
```

### diagnostics

```python
from PIL import Image


def _read_json_safe(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }


def collect_sd3_intervention_diffs(stage_dir: Path, *, image_name: str, kind: str) -> pd.DataFrame:
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = CAPTURE_DIR / "samples" / base_id / "image.png"
        intervention_path = stage_dir / "samples" / sample_id / image_name
        if not baseline_path.exists() or not intervention_path.exists():
            continue
        meta = {
            "kind": kind,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "target_word": row.get("target_word", ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "image_path": str(intervention_path),
            "baseline_path": str(baseline_path),
        }
        meta.update(_image_diff_metrics(baseline_path, intervention_path))
        out.append(meta)
    return pd.DataFrame(out)


diff_tables = []
if BLOCK_SCALE_DIR.exists():
    diff_tables.append(collect_sd3_intervention_diffs(BLOCK_SCALE_DIR, image_name="ablation.png", kind="block_scaling_strong"))

if SAE_LABEL_PROBE_ROOT.exists():
    for label_dir in sorted(SAE_LABEL_PROBE_ROOT.iterdir()):
        if label_dir.is_dir():
            diff_tables.append(collect_sd3_intervention_diffs(label_dir, image_name="steered.png", kind=f"sae_label/{label_dir.name}"))

all_diffs = pd.concat([x for x in diff_tables if len(x)], ignore_index=True) if diff_tables else pd.DataFrame()
STAGE5_ROOT.mkdir(parents=True, exist_ok=True)

if len(all_diffs):
    all_diffs.to_csv(STAGE5_DIFFS_CSV, index=False)
    print("Saved:", STAGE5_DIFFS_CSV)
    display(
        all_diffs.groupby("kind")
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_median=("l1_mean", "median"),
            rmse_mean=("rmse", "mean"),
            changed_px_005=("changed_px_005", "mean"),
            changed_px_010=("changed_px_010", "mean"),
            changed_px_020=("changed_px_020", "mean"),
        )
        .sort_values("l1_mean", ascending=False)
    )
    print("Most changed examples:")
    display(all_diffs.sort_values("l1_mean", ascending=False).head(20)[["kind", "sample_id", "beta", "scale", "l1_mean", "rmse", "changed_px_010"]])
    print("Least changed examples:")
    display(all_diffs.sort_values("l1_mean", ascending=True).head(20)[["kind", "sample_id", "beta", "scale", "l1_mean", "rmse", "changed_px_010"]])
else:
    print("No intervention images found for diff diagnostics yet.")
```

### 5.1 Source hotfix for SD3 image-token steering


```python
PATCH_STAGE5_SOURCE_HOTFIXES = True


def _patch_between_markers(path: Path, start_marker: str, end_marker: str, new_text: str, patch_marker: str):
    text = path.read_text(encoding="utf-8")
    if patch_marker in text:
        print(f"Already patched: {path.name} ({patch_marker})")
        return False
    start = text.index(start_marker)
    end = text.index(end_marker, start)
    backup = path.with_suffix(path.suffix + f".before_{patch_marker.lower()}")
    if not backup.exists():
        backup.write_text(text, encoding="utf-8")
    path.write_text(text[:start] + new_text.rstrip() + text[end:], encoding="utf-8")
    print("Patched:", path)
    print("Backup:", backup)
    return True

if PATCH_STAGE5_SOURCE_HOTFIXES:
    act_path = PROJECT_ROOT / "src" / "diffusion_attention_analysis_v2" / "activations.py"
    act_text = act_path.read_text(encoding="utf-8")

    if "SD3_TUPLE_SPATIAL_HOTFIX" not in act_text:
        new_extract = """def extract_first_tensor(obj: Any):
    # SD3_TUPLE_SPATIAL_HOTFIX:
    # Diffusers SD3 JointTransformerBlock can return a tuple like
    # (encoder_hidden_states, hidden_states). The first tensor is the text/context
    # stream, while the second tensor is the image-token stream needed for mask
    # alignment and SAE steering. For such tuples, choose the tensor with more tokens.
    if isinstance(obj, torch.Tensor):
        return obj
    if isinstance(obj, (tuple, list)):
        direct_tensors = [item for item in obj if isinstance(item, torch.Tensor)]
        if len(direct_tensors) >= 2:
            a, b = direct_tensors[0], direct_tensors[1]
            if a.ndim == 3 and b.ndim == 3 and b.shape[1] > a.shape[1]:
                return b
        for item in obj:
            found = extract_first_tensor(item)
            if found is not None:
                return found
    if isinstance(obj, dict):
        for value in obj.values():
            found = extract_first_tensor(value)
            if found is not None:
                return found
    return None
"""
        _patch_between_markers(
            act_path,
            "def extract_first_tensor",
            "\n\ndef replace_first_tensor",
            new_extract,
            "SD3_TUPLE_SPATIAL_HOTFIX",
        )
    else:
        print("extract_first_tensor already has SD3 spatial hotfix")

    act_text = act_path.read_text(encoding="utf-8")
    if "SD3_RECURSIVE_REPLACE_HOTFIX" not in act_text:
        new_replace = """def replace_first_tensor(obj: Any, new_tensor: torch.Tensor, old_shape) -> Any:
    # SD3_RECURSIVE_REPLACE_HOTFIX:
    # Recursively replace the first tensor matching old_shape. This is safer for
    # tuple outputs such as (encoder_hidden_states, hidden_states) and nested structures.
    def _replace(x: Any):
        if isinstance(x, torch.Tensor):
            if tuple(x.shape) == tuple(old_shape):
                return new_tensor, True
            return x, False
        if isinstance(x, tuple):
            out = []
            changed_any = False
            for item in x:
                if changed_any:
                    out.append(item)
                    continue
                new_item, changed = _replace(item)
                out.append(new_item)
                changed_any = changed_any or changed
            return tuple(out), changed_any
        if isinstance(x, list):
            out = []
            changed_any = False
            for item in x:
                if changed_any:
                    out.append(item)
                    continue
                new_item, changed = _replace(item)
                out.append(new_item)
                changed_any = changed_any or changed
            return out, changed_any
        if isinstance(x, dict):
            out = {}
            changed_any = False
            for key, value in x.items():
                if changed_any:
                    out[key] = value
                    continue
                new_value, changed = _replace(value)
                out[key] = new_value
                changed_any = changed_any or changed
            return out, changed_any
        return x, False

    out, _ = _replace(obj)
    return out
"""
        _patch_between_markers(
            act_path,
            "def replace_first_tensor",
            "\n\ndef flatten_feature_map",
            new_replace,
            "SD3_RECURSIVE_REPLACE_HOTFIX",
        )
    else:
        print("replace_first_tensor already has recursive hotfix")

    int_path = PROJECT_ROOT / "src" / "diffusion_attention_analysis_v2" / "interventions.py"
    int_text = int_path.read_text(encoding="utf-8")
    if "SD3_DECODER_DIRECTION_STEERING_HOTFIX" not in int_text:
        new_class = """class ResidualConceptSteerer:
    # SD3_DECODER_DIRECTION_STEERING_HOTFIX:
    # Add SAE decoder concept directions to the original residual update.
    # This preserves the reconstruction residual and avoids replacing the block output
    # with an imperfect SAE reconstruction. For SD3 CFG batches, default branch mode
    # is cond-only and can be changed via SD3_SAE_BRANCH_MODE={cond,uncond,all}.
    def __init__(self, sae, concept_ids: Sequence[int], beta: float, *, step_getter: Callable[[], int], active_steps: Sequence[int] | None = None, device: str = "cuda") -> None:
        self.sae = sae.to(device)
        self.concept_ids = [int(x) for x in concept_ids]
        self.beta = float(beta)
        self.step_getter = step_getter
        self.active_steps = {int(x) for x in active_steps or []}
        self.device = device
        import os as _os
        self.branch_mode = _os.environ.get("SD3_SAE_BRANCH_MODE", "cond").strip().lower()
        self.direction_mode = _os.environ.get("SD3_SAE_DIRECTION_MODE", "mean_decoder_direction").strip().lower()

    def _direction(self, flat: torch.Tensor) -> torch.Tensor | None:
        with torch.no_grad():
            # decoder.weight has shape [d_model, d_hidden]; columns are concept vectors.
            W = self.sae.decoder.weight.detach().to(device=flat.device, dtype=flat.dtype)
            valid = [i for i in self.concept_ids if 0 <= i < W.shape[1]]
            if not valid:
                return None
            direction = W[:, valid]
            if self.direction_mode in {"sum", "sum_decoder_direction"}:
                direction = direction.sum(dim=1)
            else:
                direction = direction.mean(dim=1)
            return direction.reshape(1, -1)

    def _apply_branch_mask(self, delta_flat: torch.Tensor, feature_shape: Sequence[int]) -> torch.Tensor:
        if len(feature_shape) != 3:
            return delta_flat
        b, n, c = [int(x) for x in feature_shape]
        if b < 2 or b % 2 != 0:
            return delta_flat
        branch_mode = self.branch_mode
        if branch_mode not in {"cond", "conditional", "uncond", "unconditional", "all"}:
            branch_mode = "cond"
        if branch_mode == "all":
            return delta_flat
        delta = torch.zeros_like(delta_flat).reshape(b, n, c)
        d = delta_flat.reshape(b, n, c)
        half = b // 2
        if branch_mode in {"cond", "conditional"}:
            delta[half:] = d[half:]
        elif branch_mode in {"uncond", "unconditional"}:
            delta[:half] = d[:half]
        return delta.reshape_as(delta_flat)

    def __call__(self, module, inputs, output):
        step = int(self.step_getter())
        if self.active_steps and step not in self.active_steps:
            return output
        x_in = extract_first_tensor(inputs)
        x_out = extract_first_tensor(output)
        if x_out is None:
            return output
        residual_mode = x_in is not None and tuple(x_in.shape) == tuple(x_out.shape)
        update = x_out - x_in if residual_mode else x_out
        try:
            flat, shape = flatten_feature_map(update.detach().float())
        except ValueError:
            return output
        flat = flat.to(self.device)
        direction = self._direction(flat)
        if direction is None:
            return output
        with torch.no_grad():
            delta = self.beta * direction.expand_as(flat)
            delta = self._apply_branch_mask(delta, shape)
            steered_flat = flat + delta
        steered = unflatten_feature_map(steered_flat, shape, like=x_out)
        new_tensor = x_in + steered if residual_mode else steered
        return replace_first_tensor(output, new_tensor, x_out.shape)
"""
        _patch_between_markers(
            int_path,
            "class ResidualConceptSteerer:",
            "\n\nclass BlockOutputScaler:",
            new_class,
            "SD3_DECODER_DIRECTION_STEERING_HOTFIX",
        )
    else:
        print("ResidualConceptSteerer already has decoder-direction hotfix")

os.environ["SD3_SAE_BRANCH_MODE"] = "cond"
os.environ["SD3_SAE_DIRECTION_MODE"] = "mean_decoder_direction"
print("SD3_SAE_BRANCH_MODE:", os.environ["SD3_SAE_BRANCH_MODE"])
print("SD3_SAE_DIRECTION_MODE:", os.environ["SD3_SAE_DIRECTION_MODE"])
```

### 5.2 Specificity-aware concept selection and probe prompts


```python
def _yaml_list(xs):
    return "[" + ",".join(str(x) for x in xs) + "]"


def _safe_file_part(text: str) -> str:
    text = str(text).strip().lower()
    text = re.sub(r"[^a-z0-9_-]+", "_", text)
    text = re.sub(r"_+", "_", text).strip("_")
    return text or "label"


def _target_word_from_record(row: dict) -> str:
    if row.get("target_word"):
        return str(row["target_word"]).strip().lower()
    for key in ("target", "label"):
        if row.get(key):
            return str(row[key]).strip().lower()
    entities = row.get("entities") or []
    if entities:
        ent = entities[0]
        if isinstance(ent, dict):
            return str(ent.get("token") or ent.get("label") or ent.get("name") or "").strip().lower()
        return str(ent).strip().lower()
    return ""


def _read_jsonl(path: Path) -> list[dict]:
    rows = []
    with path.open("r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if line:
                rows.append(json.loads(line))
    return rows


PROBE_PROMPTS_PER_LABEL_V2 = 3
MAX_STAGE5_LABELS_V2 = 6
FEATURES_PER_LABEL_V2 = 2
MAX_LABELS_PER_FEATURE_STRICT = 6
MAX_LABELS_PER_FEATURE_RELAXED = 12
MIN_LABEL_SHARE_STRICT = 0.16
MIN_SPECIFICITY_MARGIN_STRICT = 1.0
ALLOW_BROAD_FEATURE_FALLBACK = False

PREFERRED_PROBE_LABELS_V2 = [
    "dog", "cat", "car", "bicycle", "cup", "book", "vase", "flower", "person", "umbrella",
    "tree", "bench", "chair", "table", "bird", "horse", "boat", "laptop", "backpack", "bottle",
    "apple", "banana", "clock", "train", "bus", "sofa", "bed", "window", "skateboard", "airplane",
]

concept_csv = CONCEPT_DIR / "concept_dictionary.csv"
concept_df = pd.read_csv(concept_csv)
assert len(concept_df), concept_csv
concept_df["label"] = concept_df["label"].astype(str).str.lower()
concept_df["feature_id"] = pd.to_numeric(concept_df["feature_id"], errors="coerce").astype("Int64")
concept_df["score"] = pd.to_numeric(concept_df.get("score", 0.0), errors="coerce").fillna(0.0)
concept_df = concept_df.dropna(subset=["feature_id"]).copy()
concept_df["feature_id"] = concept_df["feature_id"].astype(int)

fl = (
    concept_df.groupby(["feature_id", "label"], as_index=False)
    .agg(score=("score", "max"), count=("count", "max"))
)
feature_totals = fl.groupby("feature_id").agg(
    n_labels=("label", "nunique"),
    sum_score=("score", "sum"),
    max_score_any=("score", "max"),
).reset_index()
fl = fl.merge(feature_totals, on="feature_id", how="left")
fl["label_share"] = fl["score"] / fl["sum_score"].clip(lower=1e-8)

# Compute best competing label score for each feature.
other_max_rows = []
for fid, grp in fl.groupby("feature_id"):
    grp_sorted = grp.sort_values("score", ascending=False).reset_index(drop=True)
    for idx, row in grp_sorted.iterrows():
        other_max = float(grp_sorted.loc[1, "score"]) if idx == 0 and len(grp_sorted) > 1 else float(grp_sorted.loc[0, "score"])
        rank = idx + 1
        other_max_rows.append({"feature_id": int(fid), "label": row["label"], "other_max_score": other_max, "label_rank_for_feature": rank})
other_df = pd.DataFrame(other_max_rows)
fl = fl.merge(other_df, on=["feature_id", "label"], how="left")
fl["specificity_margin"] = fl["score"] - fl["other_max_score"].fillna(0.0)
fl["selectivity_score"] = fl["score"] * fl["label_share"] / np.sqrt(fl["n_labels"].clip(lower=1))

strict = (
    (fl["n_labels"] <= MAX_LABELS_PER_FEATURE_STRICT)
    & ((fl["specificity_margin"] >= MIN_SPECIFICITY_MARGIN_STRICT) | (fl["label_share"] >= MIN_LABEL_SHARE_STRICT))
)
relaxed = (
    (fl["n_labels"] <= MAX_LABELS_PER_FEATURE_RELAXED)
    & ((fl["specificity_margin"] >= 0.25) | (fl["label_share"] >= 0.10))
)
fl["selection_tier"] = np.where(strict, "strict", np.where(relaxed, "relaxed", "reject"))

# Save the worst reusable features for audit.
rejected_or_broad = (
    fl.sort_values(["n_labels", "score"], ascending=[False, False])
    .groupby("feature_id")
    .head(1)
    .sort_values(["n_labels", "max_score_any"], ascending=False)
)
rejected_or_broad.to_csv(STAGE5_REJECTED_FEATURES_CSV, index=False)

prompt_records_all = _read_jsonl(PROMPTS_150)
for row in prompt_records_all:
    row["target_word"] = _target_word_from_record(row)

target_counts = pd.Series([r["target_word"] for r in prompt_records_all if r.get("target_word")]).value_counts().to_dict()
concept_labels = set(concept_df["label"].dropna().unique().tolist())


def select_concepts_for_label(label: str, k: int = FEATURES_PER_LABEL_V2, *, allow_reject_fallback: bool = ALLOW_BROAD_FEATURE_FALLBACK) -> pd.DataFrame:
    rows = fl[fl["label"] == label].copy()
    if not len(rows):
        return rows
    tier_order = {"strict": 0, "relaxed": 1, "reject": 2}
    rows["tier_order"] = rows["selection_tier"].map(tier_order).fillna(3)
    ok = rows[rows["selection_tier"].isin(["strict", "relaxed"])]
    if len(ok):
        rows = ok
    elif not allow_reject_fallback:

        return rows.iloc[0:0]
    rows = rows.sort_values(
        ["tier_order", "selectivity_score", "specificity_margin", "score"],
        ascending=[True, False, False, False],
    )
    return rows.head(k)

candidate_labels = [
    label for label in PREFERRED_PROBE_LABELS_V2
    if target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL_V2 and label in concept_labels and len(select_concepts_for_label(label)) > 0
]

if len(candidate_labels) < MAX_STAGE5_LABELS_V2:
    ranked_labels = (
        fl[fl["selection_tier"].isin(["strict", "relaxed"])]
        .groupby("label", as_index=False)
        .agg(best_selectivity=("selectivity_score", "max"), best_score=("score", "max"))
        .sort_values(["best_selectivity", "best_score"], ascending=False)["label"]
        .tolist()
    )
    for label in ranked_labels:
        if label not in candidate_labels and target_counts.get(label, 0) >= PROBE_PROMPTS_PER_LABEL_V2:
            candidate_labels.append(label)
        if len(candidate_labels) >= MAX_STAGE5_LABELS_V2:
            break

TARGET_LABELS_FOR_PROBE_V2 = candidate_labels[:MAX_STAGE5_LABELS_V2]
assert TARGET_LABELS_FOR_PROBE_V2, (
    "No labels passed the specificity filter. First inspect stage5_v2_rejected_feature_reuse.csv; "
    "then either relax MAX_LABELS_PER_FEATURE_* / MIN_LABEL_SHARE_* or set ALLOW_BROAD_FEATURE_FALLBACK=True for a diagnostic-only run."
)
print("Stage 5 v2 probe labels:", TARGET_LABELS_FOR_PROBE_V2)

PROBE_PROMPT_ROOT_V2.mkdir(parents=True, exist_ok=True)
probe_records_v2 = []
probe_label_prompt_paths_v2 = {}
label_to_concept_ids_v2 = {}
selection_rows = []

for label in TARGET_LABELS_FOR_PROBE_V2:
    selected = select_concepts_for_label(label, FEATURES_PER_LABEL_V2)
    concept_ids = selected["feature_id"].astype(int).tolist()
    if not concept_ids:
        print("[skip] no concept ids", label)
        continue
    label_to_concept_ids_v2[label] = concept_ids
    for _, row in selected.iterrows():
        selection_rows.append({
            "label": label,
            "feature_id": int(row["feature_id"]),
            "score": float(row["score"]),
            "n_labels": int(row["n_labels"]),
            "label_share": float(row["label_share"]),
            "specificity_margin": float(row["specificity_margin"]),
            "selectivity_score": float(row["selectivity_score"]),
            "selection_tier": str(row["selection_tier"]),
            "label_rank_for_feature": int(row["label_rank_for_feature"]),
        })

    label_records = [dict(r) for r in prompt_records_all if r.get("target_word") == label][:PROBE_PROMPTS_PER_LABEL_V2]
    for r in label_records:
        r["target_word"] = label

    prompt_path = PROBE_PROMPT_ROOT_V2 / f"sd3_v2_probe_{_safe_file_part(label)}.jsonl"
    with prompt_path.open("w", encoding="utf-8") as f:
        for r in label_records:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")
    probe_label_prompt_paths_v2[label] = prompt_path
    probe_records_v2.extend(label_records)

selection_df = pd.DataFrame(selection_rows)
selection_df.to_csv(STAGE5_FEATURE_SELECTION_CSV, index=False)

PROBE_PROMPTS_ALL_V2 = PROBE_PROMPT_ROOT_V2 / "sd3_v2_probe_all_labels.jsonl"
with PROBE_PROMPTS_ALL_V2.open("w", encoding="utf-8") as f:
    for r in probe_records_v2:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

BLOCK_PROBE_PROMPTS_V2 = PROBE_PROMPT_ROOT_V2 / "sd3_v2_block_probe.jsonl"
with BLOCK_PROBE_PROMPTS_V2.open("w", encoding="utf-8") as f:
    for r in probe_records_v2[: min(12, len(probe_records_v2))]:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

print("Selected concept IDs by label:")
display(selection_df)
print("Most broadly reused features, saved for audit:", STAGE5_REJECTED_FEATURES_CSV)
display(rejected_or_broad.head(20)[["feature_id", "label", "score", "n_labels", "label_share", "specificity_margin", "selection_tier"]])

prompt_summary = pd.DataFrame([
    {
        "label": label,
        "n_prompts": sum(1 for r in probe_records_v2 if r.get("target_word") == label),
        "concept_ids": label_to_concept_ids_v2.get(label, []),
        "prompt_file": str(probe_label_prompt_paths_v2.get(label)),
    }
    for label in TARGET_LABELS_FOR_PROBE_V2
])
display(prompt_summary)
print("Combined SAE prompts:", PROBE_PROMPTS_ALL_V2)
print("Block prompts:", BLOCK_PROBE_PROMPTS_V2)
```

```python
RUN_SD3_BLOCK_SCALING_V2 = True
BLOCK_V2_MODULES = ["transformer_blocks.0", "transformer_blocks.11", "transformer_blocks.23"]
BLOCK_V2_WINDOWS = {
    "early": [0, 1, 2, 3, 4],
    "mid": [11, 12, 13, 14, 15],
    "late": [23, 24, 25, 26, 27],
}
BLOCK_V2_RUN_WINDOWS = ["early", "mid", "late"]
BLOCK_V2_SCALES = [0.75, 1.25]
BLOCK_V2_MAX_PROMPTS = min(4, len(probe_records_v2))

block_v2_reports = []
for module in BLOCK_V2_MODULES:
    for window_name in BLOCK_V2_RUN_WINDOWS:
        target_steps = BLOCK_V2_WINDOWS[window_name]
        out_dir = BLOCK_SOFT_ROOT / f"{_safe_file_part(module)}__{window_name}"
        print(f"\n=== block scaling v2 | module={module} | window={window_name} | steps={target_steps} ===")
        overrides = [
            f"data.prompts_path={BLOCK_PROBE_PROMPTS_V2}",
            f"data.output_dir={out_dir}",
            f"data.max_prompts={BLOCK_V2_MAX_PROMPTS}",
            f"runtime.seed_start={SEED_START}",
            f"runtime.num_seeds={NUM_SEEDS}",
            f"intervention.modules={_yaml_list([module])}",
            f"intervention.target_steps={_yaml_list(target_steps)}",
            f"intervention.scales={_yaml_list(BLOCK_V2_SCALES)}",
        ]
        run_config(str(SD3_BLOCK_CFG_PATH_V2), overrides=overrides, dry_run=True)
        if RUN_SD3_BLOCK_SCALING_V2:
            run_config(str(SD3_BLOCK_CFG_PATH_V2), overrides=overrides, dry_run=False)
            report = show_report(str(out_dir / "report.json")) or {}
            block_v2_reports.append({"module": module, "window": window_name, "out_dir": str(out_dir), **report})

if block_v2_reports:
    display(pd.DataFrame(block_v2_reports))
else:
    print("RUN_SD3_BLOCK_SCALING_V2=False: dry-runs only.")
```

### 5.5 Stage 5b v2 — specificity-aware SAE steering near the trained step

```python
RUN_SD3_SAE_SPECIFICITY_V2 = True
SAE_V2_WINDOW_NAME = "mid"
SAE_V2_TARGET_STEPS = [max(0, SAE_STEP - 2), max(0, SAE_STEP - 1), SAE_STEP, SAE_STEP + 1, SAE_STEP + 2]
SAE_V2_BETAS = [-5.0, 0.0, 5.0, 10.0]
SAE_V2_MAX_PROMPTS_PER_LABEL = PROBE_PROMPTS_PER_LABEL_V2

os.environ["SD3_SAE_BRANCH_MODE"] = "cond"
os.environ["SD3_SAE_DIRECTION_MODE"] = "mean_decoder_direction"

assert (SAE_DIR / "sae.pt").exists(), f"Missing trained spatial SAE checkpoint: {SAE_DIR / 'sae.pt'}"
assert label_to_concept_ids_v2, "No selected concept IDs. Run the specificity selection cell first."

sae_v2_reports = []
for label in TARGET_LABELS_FOR_PROBE_V2:
    prompt_file = probe_label_prompt_paths_v2.get(label)
    concept_ids = label_to_concept_ids_v2.get(label, [])
    if not prompt_file or not concept_ids:
        print(f"[skip] {label}: prompt_file={prompt_file}, concept_ids={concept_ids}")
        continue

    out_dir = SAE_SPECIFIC_ROOT / _safe_file_part(label) / SAE_V2_WINDOW_NAME
    print(f"\n=== SAE specificity v2: {label} | concept_ids={concept_ids} | steps={SAE_V2_TARGET_STEPS} ===")

    overrides = [
        f"data.prompts_path={prompt_file}",
        f"data.sae_checkpoint={SAE_DIR / 'sae.pt'}",
        f"data.output_dir={out_dir}",
        f"data.max_prompts={SAE_V2_MAX_PROMPTS_PER_LABEL}",
        f"runtime.seed_start={SEED_START}",
        f"runtime.num_seeds={NUM_SEEDS}",
        f"intervention.module={SAE_LAYER}",
        f"intervention.concept_ids={_yaml_list(concept_ids)}",
        f"intervention.betas={_yaml_list(SAE_V2_BETAS)}",
        f"intervention.target_steps={_yaml_list(SAE_V2_TARGET_STEPS)}",
    ]

    run_config(str(SD3_SAE_CFG_PATH_V2), overrides=overrides, dry_run=True)
    if RUN_SD3_SAE_SPECIFICITY_V2:
        run_config(str(SD3_SAE_CFG_PATH_V2), overrides=overrides, dry_run=False)
        report = show_report(str(out_dir / "report.json")) or {}
        sae_v2_reports.append({"label": label, "window": SAE_V2_WINDOW_NAME, "concept_ids": concept_ids, "out_dir": str(out_dir), **report})

if sae_v2_reports:
    display(pd.DataFrame(sae_v2_reports))
else:
    print("RUN_SD3_SAE_SPECIFICITY_V2=False: dry-runs only.")
```

### 5.7 Stage 5 v2 diagnostics


```python
from PIL import Image


def _read_json_safe(path: Path, default=None):
    if not path.exists():
        return default
    return json.loads(path.read_text(encoding="utf-8"))


def _base_sample_id_from_intervention(sample_id: str) -> str | None:
    if "_seed_" not in sample_id:
        return None
    prefix, rest = sample_id.split("_seed_", 1)
    seed = rest.split("_", 1)[0]
    return f"{prefix}_seed_{seed}"


def _find_baseline_image(base_id: str) -> Path | None:
    candidates = [
        CAPTURE_DIR / "samples" / base_id / "image.png",
        SPATIAL_CAPTURE_DIR / "samples" / base_id / "image.png",
        ORIGINAL_CAPTURE_DIR / "samples" / base_id / "image.png",
    ]
    for p in candidates:
        if p.exists():
            return p
    return None


def _image_diff_metrics(a_path: Path, b_path: Path, size=(256, 256)) -> dict:
    a = Image.open(a_path).convert("RGB").resize(size)
    b = Image.open(b_path).convert("RGB").resize(size)
    x = np.asarray(a).astype(np.float32) / 255.0
    y = np.asarray(b).astype(np.float32) / 255.0
    diff = np.abs(y - x)
    diff_gray = diff.mean(axis=2)
    return {
        "l1_mean": float(diff.mean()),
        "rmse": float(np.sqrt(((y - x) ** 2).mean())),
        "changed_px_005": float((diff_gray > 0.05).mean()),
        "changed_px_010": float((diff_gray > 0.10).mean()),
        "changed_px_020": float((diff_gray > 0.20).mean()),
    }


def collect_stage5_v2_diffs(stage_dir: Path, *, image_name: str, kind: str, label: str | None = None, module: str | None = None, window: str | None = None) -> pd.DataFrame:
    rows = _read_json_safe(stage_dir / "intervention_rows.json", []) or []
    out = []
    for row in rows:
        sample_id = row.get("sample_id")
        if not sample_id:
            continue
        base_id = _base_sample_id_from_intervention(sample_id)
        if not base_id:
            continue
        baseline_path = _find_baseline_image(base_id)
        intervention_path = stage_dir / "samples" / sample_id / image_name
        if baseline_path is None or not intervention_path.exists():
            continue
        meta = {
            "kind": kind,
            "label": label or row.get("target_word", ""),
            "module": module or (row.get("requested_modules") or [None])[0],
            "window": window,
            "stage_dir": str(stage_dir),
            "sample_id": sample_id,
            "base_id": base_id,
            "prompt": row.get("prompt", ""),
            "target_word": row.get("target_word", label or ""),
            "beta": row.get("beta", np.nan),
            "scale": row.get("scale", np.nan),
            "concept_ids": row.get("concept_ids", []),
            "edited_path": str(intervention_path),
            "baseline_path": str(baseline_path),
        }
        meta.update(_image_diff_metrics(baseline_path, intervention_path))
        out.append(meta)
    return pd.DataFrame(out)


diff_tables = []

if BLOCK_SOFT_ROOT.exists():
    for d in sorted([p for p in BLOCK_SOFT_ROOT.iterdir() if p.is_dir()]):
        name = d.name
        if "__" in name:
            module_part, window = name.rsplit("__", 1)
            module = module_part.replace("transformer_blocks_", "transformer_blocks.")
        else:
            module, window = None, None
        diff_tables.append(collect_stage5_v2_diffs(d, image_name="ablation.png", kind="block_soft_v2", module=module, window=window))

if SAE_SPECIFIC_ROOT.exists():
    for label_dir in sorted([p for p in SAE_SPECIFIC_ROOT.iterdir() if p.is_dir()]):
        for window_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
            diff_tables.append(collect_stage5_v2_diffs(window_dir, image_name="steered.png", kind="sae_specificity_v2", label=label_dir.name, window=window_dir.name))

if SAE_TEMPORAL_ROOT.exists():
    for label_dir in sorted([p for p in SAE_TEMPORAL_ROOT.iterdir() if p.is_dir()]):
        for window_dir in sorted([p for p in label_dir.iterdir() if p.is_dir()]):
            diff_tables.append(collect_stage5_v2_diffs(window_dir, image_name="steered.png", kind="sae_temporal_v2", label=label_dir.name, window=window_dir.name))

all_diffs_v2 = pd.concat([x for x in diff_tables if len(x)], ignore_index=True) if diff_tables else pd.DataFrame()

if len(all_diffs_v2):
    all_diffs_v2["artifact_flag"] = (all_diffs_v2["l1_mean"] > 0.35) | (all_diffs_v2["changed_px_020"] > 0.70)
    all_diffs_v2["moderate_change"] = (all_diffs_v2["l1_mean"].between(0.03, 0.25)) & (~all_diffs_v2["artifact_flag"])
    all_diffs_v2.to_csv(STAGE5_V2_DIFFS_CSV, index=False)
    print("Saved:", STAGE5_V2_DIFFS_CSV)

    print("Overall summary:")
    display(
        all_diffs_v2.groupby(["kind", "window"], dropna=False)
        .agg(
            n=("l1_mean", "count"),
            l1_mean=("l1_mean", "mean"),
            l1_median=("l1_mean", "median"),
            changed_px_010=("changed_px_010", "mean"),
            changed_px_020=("changed_px_020", "mean"),
            artifact_rate=("artifact_flag", "mean"),
            moderate_rate=("moderate_change", "mean"),
        )
        .sort_values(["kind", "l1_mean"], ascending=[True, False])
    )

    if (all_diffs_v2["kind"] == "block_soft_v2").any():
        print("Block scaling by module/window/scale:")
        display(
            all_diffs_v2[all_diffs_v2["kind"] == "block_soft_v2"]
            .groupby(["module", "window", "scale"], dropna=False)
            .agg(n=("l1_mean", "count"), l1_mean=("l1_mean", "mean"), artifact_rate=("artifact_flag", "mean"))
            .reset_index()
            .sort_values("l1_mean", ascending=False)
        )

    if (all_diffs_v2["kind"] == "sae_specificity_v2").any():
        print("SAE steering by label/beta:")
        display(
            all_diffs_v2[all_diffs_v2["kind"] == "sae_specificity_v2"]
            .groupby(["label", "window", "beta"], dropna=False)
            .agg(n=("l1_mean", "count"), l1_mean=("l1_mean", "mean"), changed_px_020=("changed_px_020", "mean"), artifact_rate=("artifact_flag", "mean"), moderate_rate=("moderate_change", "mean"))
            .reset_index()
            .sort_values(["label", "beta"])
        )

    print("Sanity check beta=0 rows, should be close to zero if deterministic and additive patch is active:")
    beta0 = all_diffs_v2[(all_diffs_v2["kind"].str.startswith("sae")) & (pd.to_numeric(all_diffs_v2["beta"], errors="coerce") == 0)]
    display(beta0[["label", "sample_id", "l1_mean", "changed_px_010", "artifact_flag"]].head(20))

    print("Most useful moderate candidates:")
    display(
        all_diffs_v2[all_diffs_v2["moderate_change"]]
        .sort_values(["kind", "l1_mean"], ascending=[True, False])
        .head(30)[["kind", "label", "module", "window", "sample_id", "beta", "scale", "l1_mean", "changed_px_010", "changed_px_020", "artifact_flag"]]
    )
else:
    print("No Stage 5 v2 images found yet. Run Stage 5a/5b v2 cells first.")
```
